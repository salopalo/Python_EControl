## Exercicio 1

Dada a seguinte lista na que se indica a cotización media mensual do Euríbor ao longo do ano 2025:

 **euríbor = [2.525, 2.407, 2.398, 2.143, 2.081, 2.081, 2.079, 2.114, 2.172, 2.187, 2.217]** 

Calcula mediante un **"for"** e unha variable chamada **"total"** (que faga de acumulador) a media do segundo trimestre (abril-xuño) e mostra o resultado por pantalla.

```python
euríbor = [2.525, 2.407, 2.398, 2.143, 2.081, 2.081, 2.079, 2.114, 2.172, 2.187, 2.217]
total = 0
for DATO in euríbor[3:6]:
    total += DATO
print(f"A media do segundo trimestre é: {total/3:.3f}")
```

 **euríbor = [2.525, 2.407, 2.398, 2.143, 2.081, 2.081, 2.079, 2.114, 2.172, 2.187, 2.217]** 

Calcula mediante ***"slicing"*** e máis as funcións intrínsecas **sum** e **len** a media do terceiro trimestre (xullo-setembro) e mostra o resultado por pantalla.

Para crear o diccionario por comprensión utiliza as dúas listas: **euríbor** e **meses_inglés** e un **"for"** que itere desde 0 ata 11.

## Exercicio 5

Utilizando o diccionario creado no exercicio anterior:

**euríbor_2025 = {'Jan': 2.525, 'Feb': 2.407, 'Mar': 2.398, 'Apr': 2.143, 'May': 2.081, 'Jun': 2.081, 'Jul': 2.079, 'Aug': 2.114, 'Sep': 2.172, 'Oct': 2.187, 'Nov': 2.217, 'Dec': 2.271}**

Mostra por pantalla o nome do mes (abreviado en inglés) que teña por euríbor o dato 2.114. Para atopalo recorre o diccionario cun **"for"** ata dar co primeiro valor que coincida co dato.

Utilizando a función ***max()*** mostra por pantalla o mes no que o euríbor estivo máis alto e o seu valor.

```python
euríbor_2025 = {'Jan': 2.525, 'Feb': 2.407, 'Mar': 2.398, 'Apr': 2.143, 'May': 2.081, 'Jun': 2.081, \
                'Jul': 2.079, 'Aug': 2.114, 'Sep': 2.172, 'Oct': 2.187, 'Nov': 2.217, 'Dec': 2.271}

for clave,valor in euríbor_2025.items():
    if valor == 2.114:
        print(f"O nome do mes que ten un euríbor igual a 2.114 é: {clave}")
        break

# maximo = max(euríbor_2025.values()) Así solo compara los valores y devuelve el más alto, pero nos faltaría el mes (la clave)

posicion = max(euríbor_2025,key=euríbor_2025.get)  # Así compara los valores y devuelve la posición (la clave) del más alto
maximo = euríbor_2025[posicion]                    # Indexando el diccionario obtenemos el valor 
print(f"O mes no que o euríbor estivo máis alto foi {posicion} cun valor igual a {maximo}") 
```
