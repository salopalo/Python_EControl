# Exercicio 1

Crea unha función chamada **media_impares** que reciba un nº variable de números enteiros e devolva a media aritmética dos que sexan impares.

```python
def media_impares(*numeros):
    son_impares = [n for n in numeros if not n%2]
    return round(sum(son_impares)/len(son_impares),2)

media_impares(1,2,3,4,5,6,7,8,9)
```

# Exercicio 2

Crea unha función chamada **num_palabras_con_h** que reciba un nº variable de palabras e devolva o nº delas que conteñen a letra h.


```python
def num_palabras_con_h(*palabras):
    con_h =[p for p in palabras if "h"in p.lower()]
    return len(con_h)

num_palabras_con_h("hijo", "coche", "casa", "Hermana")
```

# Exercicio 3

Crea unha función chamada **prezo_medio_frescos** que reciba un nº variable de productos comprados e o seu prezo. Debe devolver o prezo medio dos produtos frescos (peixe, carne, legumes e verduras).


```python
def precio_medio_frescos(**produtos):
    produtos_frescos = ("peixe", "carne", "legumes", "verduras")
    son_frescos = [prezo for prod, prezo in produtos.items() if prod.lower() in produtos_frescos]
    return sum(son_frescos)/len(son_frescos)


print(precio_medio_frescos(peixe=15, carne=19, pasta=8.5, verduras=5, deterxente=3.5))
print(precio_medio_frescos(legumes=2.5, pasta=1.4, carne=18))

```

# Exercicio 4

Crea unha función chamada **media_aforo_comienzan_por_M** que reciba un nº variable de estadios e o seu aforo. Debe devolver a media do aforo dos estadios que comecen por M


```python
def media_aforo_comienzan_por_M(**estadios):
    aforos_M = [aforo for nome, aforo in estadios.items() if nome[0].upper()=="M"]
    return sum(aforos_M)/len(aforos_M)


media_aforo_comienzan_por_M(Bernabeu=85000, CampNou=105000, Metropolitano=70000, Mestalla=49000)
```

# Exercicio 5

Crea unha función chamada **conta_palabras_tamaño** que reciba un nº variable de palabras e que conte o nº delas cuxa lonxitude coincida cun argumento fixo que se lle pase ao principio.


```python
def conta_palabras_tamanio(arg, *palabras):
    conta = 0
    for p in palabras:
        if len(p) == arg:
            conta +=1
    return conta

print(conta_palabras_tamanio(5, "perro", "gato", "coche", "apartamento"))
print(conta_palabras_tamanio(5, "pero", "gato", "coche", "apartamento"))
    
```

# Exercicio 6

Crea unha función chamada **divisibles_por** que reciba un nº variable de números enteiros e que conte o nº deles que sexan divisibles por outro argumento fixo que se lle pase ao principio.


```python
def divisibles_por(arg, *numeros):
    conta = 0
    for n in numeros:
        if  n % arg == 0:
            conta +=1
    return conta

divisibles_por(2, 2,4,6, 1)

            
```

# Exercicio 7

Crea unha función chamada **produtos_prezo_menor** que reciba un nº variable de produtos comprados e o seu prezo. Debe devolver unha lista cos nomes dos produtos cuxo prezo sexa menor ao indicado noutro argumento fixo que se lle pase ao principio.


```python
def produtos_prezo_menor(prezo, **produtos):
    prezo_menor = [prod for prod, prez in produtos.items() if prez < prezo]
    return ", ".join(prezo_menor)

produtos_prezo_menor(12, mazas=7, peras=11, pescada=19, rape=21, berza=1.2)
```

# Exercicio 9

Crea unha función chamada **cotización_euríbor_semanal** que reciba un nº variable de días da semana e a cotización do euríbor en cada día. Debe devolver o valor medio de dito euríbor nos días recibidos como *kwargs*, sempre excluíndo da media os valores que coincidan con algún dos valores que se lle pasen tamén como nº variable de argumentos, pero neste caso como *args*.


```python
def cotizacion_euribor_semanal(*valores_excluintes, **valores_por_dia):
    dias_validos = {dia:valor for dia, valor in valores_por_dia.items() if valor not in valores_excluintes}
    return round(sum(dias_validos.values()) / len(dias_validos.items()), 2)

print(sum((2.28, 2.58, 2.99))/ 3)
print(cotizacion_euribor_semanal(2.56, 2.44, 2.23, lunes=2.28, martes=2.44, miercoles=2.58, jueves=2.56, viernes=2.99))



def cotizacion_euribor_semanal2(*valores_excluintes, **valores_por_dia):
    dias_validos = [float(valor) for dia, valor in valores_por_dia.items() if valor not in valores_excluintes]
    return round(sum(dias_validos) / len(dias_validos), 2)

print(sum((2.28, 2.58, 2.99))/ 3)
print(cotizacion_euribor_semanal2(2.56, 2.44, 2.23, lunes=2.28, martes=2.44, miercoles=2.58, jueves=2.56, viernes=2.99))
```
