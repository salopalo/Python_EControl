# Exercicio 1

Crea unha función chamada **media_impares** que reciba un nº variable de números enteiros e devolva a media aritmética dos que sexan impares.

```python
def media_impares(*numeros):
    son_impares = [n for n in numeros if not n%2]
    return round(sum(son_impares)/len(son_impares),2)

media_impares(1,2,3,4,5,6,7,8,9)
```

# Exercicio 2

Crea unha función chamada **num_palabras_con_h** que reciba un nº variable de palabras e devolva o nº delas que conteñen a letra h.


```python
def num_palabras_con_h(*palabras):
    con_h =[p for p in palabras if "h"in p.lower()]
    return len(con_h)

num_palabras_con_h("hijo", "coche", "casa", "Hermana")
```

# Exercicio 3

Crea unha función chamada **prezo_medio_frescos** que reciba un nº variable de productos comprados e o seu prezo. Debe devolver o prezo medio dos produtos frescos (peixe, carne, legumes e verduras).


```python
def precio_medio_frescos(**produtos):
    produtos_frescos = ("peixe", "carne", "legumes", "verduras")
    son_frescos = [prezo for prod, prezo in produtos.items() if prod.lower() in produtos_frescos]
    return sum(son_frescos)/len(son_frescos)


print(precio_medio_frescos(peixe=15, carne=19, pasta=8.5, verduras=5, deterxente=3.5))
print(precio_medio_frescos(legumes=2.5, pasta=1.4, carne=18))

```

# Exercicio 4

Crea unha función chamada **media_aforo_comienzan_por_M** que reciba un nº variable de estadios e o seu aforo. Debe devolver a media do aforo dos estadios que comecen por M


```python
def media_aforo_comienzan_por_M(**estadios):
    aforos_M = [aforo for nome, aforo in estadios.items() if nome[0].upper()=="M"]
    return sum(aforos_M)/len(aforos_M)


media_aforo_comienzan_por_M(Bernabeu=85000, CampNou=105000, Metropolitano=70000, Mestalla=49000)
```
