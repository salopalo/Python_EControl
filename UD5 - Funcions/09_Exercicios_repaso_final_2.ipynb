Crea unha función chamada **comeza_por_vogal** que reciba como argumentos:

- Un string que conteña unha palabra.

E que devolva un valor booleano indicando se a mesma comeza por vogal ou non.

Fai varias chamadas a dita función usando varios exemplos e mostrando os resultados en pantalla.

```python
def comeza_por_vogal(palabra):
    vogais = ("a", "e", "i", "o", "u", "á", "é", "í", "ó", "ú")
    if palabra[0].lower() in vogais:
        return True
    else:
        return False

print(comeza_por_vogal("Helena"))
print(comeza_por_vogal("Carlos"))    
print(comeza_por_vogal("Alegria"))
print(comeza_por_vogal("Ínclito"))
print(comeza_por_vogal("Tolai"))
print(comeza_por_vogal("érase"))
print(comeza_por_vogal("nene"))

```

# Exercicio 2

Crea unha función chamada **conta_comezan_por_vogal** que reciba como argumentos:

- Unha lista de strings contendo cada un deles unha palabra.

E que devolva o resultado de contar as palabras da lista que comezan por vogal. Deberas usar a función creada no exercicio nº1.

Fai unha chamada a dita función usando como entrada a lista ["mesa", "encerado", "árbol","pantalla", "Antonio"] e comproba o resultado.


```python
def comeza_por_vogal(palabra):
    vogais = ("a", "e", "i", "o", "u", "á", "é", "í", "ó", "ú")
    if palabra[0].lower() in vogais:
        return True
    else:
        return False
    

def conta_comezan_por_vogal(lista_palabras):
    conta = 0
    for p in lista_palabras:
        if comeza_por_vogal(p):
            conta +=1
            
    return conta

conta_comezan_por_vogal(["mesa", "encerado", "árbol","pantalla", "Antonio"])
```

# Exercicio 3

Crea unha función chamada **estación** que reciba como argumento:

- Un string que conteña unha data en formato "dd/mm/aaaa".

E que devolva outro string indicando a estación do ano na que se atoparía (primavera, verán, outono, inverno)

Fai unha chamada a dita función usando varios exemplos e mostrando o resultado por pantalla.

```python
from datetime import datetime

def estacion(data):
    fecha_ref = data[0:6]+"2026"
    fecha_datetime = datetime.strptime(fecha_ref, "%d/%m/%Y")
    if fecha_datetime > datetime.strptime("20/03/2026", "%d/%m/%Y") and fecha_datetime < datetime.strptime("21/06/2026", "%d/%m/%Y"):
        return "Primavera"
    elif fecha_datetime > datetime.strptime("20/06/2026", "%d/%m/%Y") and fecha_datetime < datetime.strptime("21/09/2026", "%d/%m/%Y"):
        return "Verano"
    elif fecha_datetime > datetime.strptime("20/09/2026", "%d/%m/%Y") and fecha_datetime < datetime.strptime("21/12/2026", "%d/%m/%Y"):
        return "Otoño"
    else:
        return "Invierno"
    
estacion("22/12/2021")
```

Crea unha función chamada **calcula_prezo_fruta** que reciba como argumentos:

- Un string indicando o tipo de fruta (peras, manzanas, plátanos,...).
- Un float que indica o peso da fruta en kg.
- O tipo de IVA que se aplicará.

E que devolva en un float o importe total da compra de dita fruta. O cálculo farase en función dos prezos sen iva rexistrados no dicionario:

**frutas = {"peras":2.75, "manzanas":3.10, "plátanos":2.20, "cerezas":5.25, "fresas":4.30}**

Fai unha chamada a dita función usando varios exemplos e mostrando o resultado por pantalla.

```python
def calcula_prezo_fruta(fruta, peso, iva):
    frutas = {"peras":2.75, "manzanas":3.10, "plátanos":2.20, "cerezas":5.25, "fresas":4.30}
    return frutas[fruta] * peso * (1 +(iva/100))

calcula_prezo_fruta("fresas", 1.645, 10)
```
