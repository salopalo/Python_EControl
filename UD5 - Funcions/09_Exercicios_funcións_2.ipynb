# Exercicio 1

Crea unha función chamada **media_parcial** que reciba como argumentos:

- Unha lista que indica a facturación dunha empresa ao longo da semana (exemplo:[530.20, 770.95, 600.24, 870.50, 685.30, 320.82, 422.90])
- Un string no que se indique: "finde" se queremos a media da fin de semana ou "diario" se queremos a media de luns a venres. Se indicamos algún valor diferente, a función devolverá o valor 0.

E que devolva o valor da media correspondente a eses días.

Fai un par de chamadas a dita función usando como exemplo a lista do enunciado e comprobando que devolve os valores 691,44 e 371,86.


```python
def media_parcial(facturacion, dias):
    if dias != "finde" and dias != "diario":
        return 0
    elif dias == "finde":
        return sum(facturacion[5:])/ len(facturacion[5:])
    else:
        return sum(facturacion[:5])/ len(facturacion[:5])
    
    
lista = [530.20, 770.95, 600.24, 870.50, 685.30, 320.82, 422.90]

print(media_parcial(lista, "finde"))
print(media_parcial(lista, "diario"))

print(media_parcial(lista, "cosa"))
```

```python
### SOLUCION


def media_parcial(factura, tipo):
    if tipo.upper()=="FINDE":
        return (sum(factura[5:7])/len(factura[5:7]))
    elif tipo.upper()=="DIARIO":
        return (sum(factura[0:5])/len(factura[0:5]))
    else:
        return 0
    
facturación = [530.20, 770.95, 600.24, 870.50, 685.30, 320.82, 422.90]

print(f"A media de facturación de luns a venres é de: {media_parcial(facturación,'diario'):.2f} euros")
print(f"A media de facturación da fin de semana é de: {media_parcial(facturación,'finde'):.2f} euros")
print(f"A media de facturación de algo distinto é de: {media_parcial(facturación,'outro'):.2f} euros")
```

Fai unha chamada a dita función usando como entrada a lista ["mesa", "encerado", "teclado", "pantalla", "televisor"] e comproba o resultado.


```python
def conta_vogais(palabra):
    if not isinstance (palabra, str):
        raise TypeError ("El dato debe ser una palabra de tipo string.")
    else:
        suma = 0
        for letra in palabra.upper():
            if letra == "A" or letra == "E" or letra =="I" or letra == "O" or letra == "U" or letra=="Á" or letra == "É" or letra == "Í" or letra =="Ó" or letra == "Ú":
                suma += 1
        return suma

def mais_vogais(lista):
    if len(lista) == 0:
        return "LISTA BALEIRA"
    else:
        cuenta_vocales = {palabra:conta_vogais(palabra) for palabra in lista}
        return (max(cuenta_vocales, key=cuenta_vocales.get))
    
lis = ["mesa", "encerado", "teclado", "pantalla", "televisor"]
lis2 = []

print(mais_vogais(lis))
print(mais_vogais(lis2))
```

```python
### SOLUCION


def conta_vogais(palabra):
    conta = 0
    for letra in palabra:
        if letra.upper() in ('A','E','I','O','U','Á','É','Í','Ó','Ú'):
            conta += 1
    return conta

def máis_vogais(lista):
    if not lista:
        return "LISTA BALEIRA"
    else:
        palabra_max = ""
        for PAL in lista:
            if conta_vogais(PAL)>=conta_vogais(palabra_max):
                palabra_max = PAL
        return palabra_max        

print(máis_vogais(["mesa", "encerado", "teclado", "pantalla", "televisor"]))
print(máis_vogais([]))
```

# Exercicio 4

Crea unha función chamada **num_días_mes** que reciba como argumento:

- Un string que conteña o nome dun mes.

E que devolva un nº enteiro indicando o número de días que ten ese mes.

Fai unha chamada a dita función usando varios exemplos e mostrando o resultado por pantalla.

```python
def num_dias(mes):
    if not mes.lower() in ("enero", "febrero", "marzo", "abril", "mayo", "junio", "julio", "agosto", "septiembre", "octubre", "noviembre", "diciembre"):
        raise ValueError ("Mes no válido.")
    elif mes.lower() in ("enero", "marzo", "mayo", "julio", "agosto", "octubre", "diciembre"):
        return 31
    elif mes.lower() in ("abril", "junio", "septiembre", "noviembre"):
        return 30
    else: return 28
    
lista = ["Enero", "Marzo", "Abril", "Febrero", "Diciembre", "MAzo"]

for mes in lista:
    print(num_dias(mes))
    
```

```python
### SOLUCION


def num_días_mes(mes):
    if mes.lower() == 'febreiro':
        return(28)
    elif mes.lower() in ('xaneiro','marzo','maio','xullo','agosto','outubro','decembro'):
        return(31)
    else:
        return(30)

print(f"O mes de abril ten {num_días_mes('Abril')} días")
print(f"O mes de febreiro ten {num_días_mes('FEBREIRO')} días")
print(f"O mes de agosto ten {num_días_mes('agosto')} días")
```

E que devolva en un float o importe total da factura.

```python
def recibo_luz(consumo, mes, tipo_iva):
    base_imponible = consumo*0.07 + num_dias(mes)*0.05 
    # return round(base_imponible + base_imponible*tipo_iva/100, 2)
    return round((consumo*0.07 + num_dias(mes)*0.05) * (1 + tipo_iva/100),2)
    # return round(base_imponible * (1+tipo_iva/100),2)

print(recibo_luz(225, "Enero", 21)) 
```

Crea unha función chamada **leva_tilde** que reciba como argumentos:

- Un string que conteña unha palabra.

E que devolva "True" ou "False" dependendo de si a palabra contén algunha vogal con tilde ou non.

Fai varias chamadas a dita función e comproba o resultado.

```python
def leva_tilde(palabra):
    lista = []
    for letra in palabra:
        if letra.lower() in ("á", "é", "í", "ó", "ú"):
            lista.append(letra)
            break
    return len(lista)>0




print(leva_tilde("casa"))
print(leva_tilde("Telón"))


def leva_tilde2(palabra):
    for letra in palabra:
        if letra.lower() in ("á", "é", "í", "ó", "ú"):
            return True
    return False



print(leva_tilde2("casaá"))
print(leva_tilde2("TeÓn"))
```

```python
### SOLUCION


def leva_tilde(palabra):
    tilde = 'False'
    for C in palabra:
        if C.lower() in ('á','é','í','ó','ú'):
            tilde = 'True'
            break
    return(tilde)

print(leva_tilde('camión'))
print(leva_tilde('paxaro'))
print(leva_tilde('Áncora'))
```

```python
### SOLUCION


def leva_tilde(palabra):
    tilde = 'False'
    for C in palabra:
        if C.lower() in ('á','é','í','ó','ú'):
            tilde = 'True'
            break
    return(tilde)

def conta_palabras_con_tilde(lista):
    conta = 0
    for PAL in lista:
        if leva_tilde(PAL)=='True':
            conta += 1
    return(conta)

print(conta_palabras_con_tilde(["café", "mesa", "pantalón", "árbore", "tambor"]))
```
