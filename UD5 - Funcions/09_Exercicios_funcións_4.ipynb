# Exercicio 2

Na chamada á función **calcula_IRPF** feita nos **Exercicios_funcións** de principio da unidade, realiza novas chamadas á función pero neste caso cambiando de orden os argumentos e nomeándoos para que faga os cálculos correctos:


```python
def calcula_IRPF(soldo,tipo,idade):
    importe = soldo * tipo / 100
    if (idade <=50):
        importe += 100
    else:
        importe -= 170
    return importe
```
