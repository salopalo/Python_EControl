# Exercicio 4

Crea unha función chamada **calc_co2e** que calcule as emisións de CO2 equivalentes dos voos e que reciba como argumentos:

- A distancia do voo en km (**dist**)
- Un valor booleano que indique voo de volta (**returnf**)
- Un valor booleano que indique se a viaxe é de 1ª clase (**firstclass**)
- Un valor que será o factor de forzamento radiativo (**radforc**) 

O **returnf** e o **firstclass** terán o valor *False* como valores por defecto, mentres que **radforc** terá *2.0* como valor por defecto.

O cálculo do valor de retorno da función farase coa fórmula: **co2e = dist * factor_de_emisión**, onde os factores de emisión son:
- 0.26744  < 700 km 
- 0.15845  700 – 2500 km
- 0.15119  > 2500 km 

Hai que ter en conta tamén que os voos de 1ª clase e os de retorno doblan as emisións de carbono.

Finalmente, as emisións multiplícanse polo factor de forzamento radiativo reflezando así o impacto total.

Realiza varias chamadas á función para comprobar o correcto funcionamento da mesma.


```python
def calc_co2e(dist, returnf=False, firstclass=False, radforc=2.0):
    if dist < 700:
        factor_de_emisión = 0.26744
    elif dist > 2500:
        factor_de_emisión = 0.15119
    else:
        factor_de_emisión = 0.15845
    emisións = dist * factor_de_emisión
    if returnf:
        emisións*=2
    if firstclass:
        emisións*=2
    emisións*= radforc
    return(emisións)

print(calc_co2e(800))
print(calc_co2e(800,returnf=True))
    
```
