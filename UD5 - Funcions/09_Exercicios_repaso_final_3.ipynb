# Exercicio 2 (Uso de *args)

Crea unha función chamada **num_palabras_con_varias_vogais** que reciba un nº variable de palabras e devolva o nº delas que conteñen máis dunha vogal.

```python
def num_palabras_con_vogais(*args):
    vogais = ("a", "e", "i", "o", "u", "á", "é", "í", "ó", "ú")
    conta_pal =0
    for palabra in args:
        conta_vogal = 0
        for l in palabra:
            if l.lower() in vogais:
                conta_vogal +=1
        if conta_vogal > 1:
            conta_pal +=1
            
    return conta_pal



l1 = ["sa", "encerado", "árbol","pantalla", "Antonio"]
print(num_palabras_con_vogais(*l1))
print(num_palabras_con_vogais("café", "marinero", "sal"))
        
```
