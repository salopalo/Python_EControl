# Exercicio 1 (Devolución de varios valores)

Crea unha función chamada **transforma_palabras** que reciba como argumento:

- Un string que conteña unha palabra.

E que devolva os seguintes valores:

- Outra palabra na que aparezan cambiadas todas as vogais 'a' por 'e' (e viceversa).
- Outra palabra na que aparezan cambiadas todas as vogais 'i' por 'u' (e viceversa).

Fai unha chamada á función usando como exemplo varias palabras. Por exemplo, para a palabra "cetrería": ("catraríe" e "cetrerúa"), ou para a palabra "tiramisú": ("tiremisú" e "turamusí").

```python
def transforma_palabras(palabra):
    aes = ("a", "á", "A", "Á")
    es = ("e", "é", "E", "É")
    ies = ("i", "í", "I", "Í")
    ues = ("u", "ú", "U", "Ú")
    
    nova_palabra = ""
    nova_palabra_2 = ""
    
    for l in palabra:
        if l in aes:
            nova_palabra += es[aes.index(l)]
        elif l in es:
            nova_palabra += aes[es.index(l)]
        else:
            nova_palabra += l
    for l in palabra:
        if l in ies:
            nova_palabra_2 += ues[ies.index(l)]
        elif l in ues:
            nova_palabra_2 += ies[ues.index(l)]
        else:
            nova_palabra_2 += l
    return (nova_palabra, nova_palabra_2)

print(
    transforma_palabras("cetrería"),
    transforma_palabras("tiramisú"))
            
```

# Exercicio 2 (Uso de *args)

Crea unha función chamada **num_palabras_con_varias_vogais** que reciba un nº variable de palabras e devolva o nº delas que conteñen máis dunha vogal.

```python
def num_palabras_con_vogais(*args):
    vogais = ("a", "e", "i", "o", "u", "á", "é", "í", "ó", "ú")
    conta_pal =0
    for palabra in args:
        conta_vogal = 0
        for l in palabra:
            if l.lower() in vogais:
                conta_vogal +=1
        if conta_vogal > 1:
            conta_pal +=1
            
    return conta_pal



l1 = ["sa", "encerado", "árbol","pantalla", "Antonio"]
print(num_palabras_con_vogais(*l1))
print(num_palabras_con_vogais("café", "marinero", "sal"))
        
```

# Exercicio 3 (Uso de *kwargs)

Crea unha función chamada **media_de_pib** que reciba un nº variable de países europeos e o seu PIB expresado en millóns de dólares americanos. Debe devolver o PIB medio dos países recibidos como argumentos.

```python
def media_de_pib(**kwargs):
    conta = 0
    suma = 0
    for pais in kwargs:
            suma += kwargs[pais]
            conta +=1
    return suma/conta

paises = {"EEUU": 125000, "China": 110000, "Japon": 85000, "Alemania": 64200}

print(
    media_de_pib(**paises), "\n",# **desempaqueta o dicionario, 
    media_de_pib(EEUU=125000, China=110000, Japon=85000, Alemania=64200))

```

# Exercicio 4 (Uso de funcións anónimas)

Dada a seguinte lista na que cada elemento é unha tupla con 3 valores: **datos = [(3, 7, 1), (11, 2, 25), (4, 12, 4), (7, 9, 2)]**, utiliza a función intrínseca **sorted()** para ordenar inversamente a mesma (de maior a menor) pola media dos dous últimos elementos de cada tupla. Deberás usar unha **función lambda** na chamada a ***sorted()***.


```python
datos = [(3, 7, 1), (11, 2, 25), (4, 12, 4), (7, 9, 2)]

print(sorted(datos, key= lambda x: (x[1]+x[2]/ 2), reverse=True))

```
