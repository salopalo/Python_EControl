# Exercicio 1

Crea unha función chamada **múltiplos_de_3_e_5_á_vez** que reciba como argumentos:

- Unha lista de números enteiros.

E que devolva outra lista distinta cos elementos que sexan múltiplos de 3 e 5 ao mesmo tempo.

Fai unha chamada a dita función usando como entrada a lista [1,3,5,11,21,30,35,45,48,50,60] e garda o seu resultado nunha variable chamada **son_múltiplos**. Imprímea por pantalla.


```python
def multiplos_de_3_e_5_a_vez(lista):
    nova_lista = []
    for i in lista:
        if i%5==0 and i%3==0:
            nova_lista.append(i)
            print(f"engadido o numero {i}")
    return nova_lista

l = [1,3,5,11,21,30,35,45,48,50,60]

multiplos_de_3_e_5_a_vez(l)
```

```python
### SOLUCION

def múltiplos_de_3_e_5_á_vez(lista):
    res = []
    for num in lista:
        if (num%3)==0 and (num%5)==0:
            res.append(num)
    return(res)

son_múltiplos = múltiplos_de_3_e_5_á_vez([1,3,5,11,21,30,35,45,48,50,60])
print(son_múltiplos)
```

# Exercicio 2

Crea unha función chamada **num_datas_posteriores** que reciba como argumentos:

- Unha lista con strings de datas en formato "dd/mm/aaaa".
- Un string cunha data de referencia (mesmo formato).

E que devolva un número enteiro que sexa o resultado de contar as datas da lista que son posteriores no tempo á data de referencia pasada como segundo argumento.

Fai varias chamadas a dita función usando distintos exemplos.


```python
from datetime import datetime

def num_datas_posteriores(lista_datas, data):
    fecha_ref = datetime.strptime(data, "%d/%m/%Y")
    cuenta = 0
    for fecha in lista_datas:
        fecha_conv = datetime.strptime(fecha, "%d/%m/%Y")
        if fecha_conv > fecha_ref:
            cuenta +=1
    return cuenta
    
l_1 = ["01/02/1978", "15/9/1999", "26/03/2010", "30/1/2012"]
num_datas_posteriores(l_1, "01/09/2000")


        
```

```python
### SOLUCION


from datetime import datetime

def num_datas_posteriores(lista_datas,data_ref):
    conta = 0
    data_referencia = datetime.strptime(data_ref,"%d/%m/%Y")
    for D in lista_datas:
        dat = datetime.strptime(D,"%d/%m/%Y")
        if dat.date() > data_referencia.date():
            conta += 1
    return(conta)

print(num_datas_posteriores(["10/02/2025","09/12/1992","30/03/2026","10/09/2023"],"05/10/2023"))


```

# Exercicio 3

Crea unha función chamada **conta_nomes_finalizan_en** que reciba como argumentos:

- Unha lista de strings con nomes de persoas.
- Unha letra que sexa a que debe pechar as palabras a buscar.

E que devolva un número enteiro que sexa o resultado de contar o número de nomes que rematan con dita letra inicial.

Fai varias chamadas a dita función usando distintos exemplos.


```python
def conta_nomes_finalizan_en(lista_nomes, letra):
    conta = 0
    for nome in lista_nomes:
        if nome[-1].lower() == letra.lower():
            conta +=1
    return conta 


l1 = ["Juana", "Ana", "Julian", "Pepe", "CArlos", "Manuel"]       

conta_nomes_finalizan_en(l1, "a")
    
```

```python
### SOLUCION


def conta_nomes_finalizan_en(lista_nomes,letra_final):
    conta = 0
    for nome in lista_nomes:
        if nome[-1].upper()==letra_final.upper():
            conta += 1
    return(conta)

print(conta_nomes_finalizan_en(['Pedro','Carla','Blanca','Luis','Javier','Rafa'],'S'))
```

# Exercicio 4

Crea unha función chamada **cifrar_nomes** que reciba como argumentos:

- Unha lista de strings con nomes de persoas.
- Un número que indica o que hai que sumar a cada caracter do nome para facer o cifrado César.

E que devolva outra lista cos nomes das persoas escritas segundo o cifrado.

Fai varias chamadas a dita función usando distintos exemplos.


```python
def cifrar_nomes(lista_nomes, numero):
    lista_2 =[]
    for nome in lista_nomes:
        palabra = ""
        for l in nome:
            palabra = palabra + chr(ord(l)+numero)
        lista_2.append(palabra)
    return lista_2

l1 = ["Juana", "Ana", "Julian", "Pepe", "CArlos", "Manuel"] 
cifrar_nomes(l1, 3)    
```

```python
### SOLUCION

def cifrar_nomes(lista_nomes,num):
    lista_cifrada = []
    for nome in lista_nomes:
        nome_cifrado = ""
        for letra in nome:
            nome_cifrado += chr(ord(letra)+num)
        lista_cifrada.append(nome_cifrado)
    return lista_cifrada

print(cifrar_nomes(['Roger','Novak','rafa','Roberto','Carlos'],3))

```
