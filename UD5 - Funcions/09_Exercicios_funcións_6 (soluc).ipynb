# Exercicio 1

Crea unha **función lambda** que reciba tres argumentos e realice a suma dos dous primeiros e divida o resultado obtido entre o terceiro argumento. A función devolverá o resultado final da operación. Asigna dita función a unha variable chamada **calcular** e fai unha chamada á mesma usando distintos exemplos e comprobando que devolve os resultados esperados.

```python
calcular = lambda x,y,z:(x+y)/z
print(calcular(4,8,2))
print(calcular(7,9,3))
```

# Exercicio 3

Dada a seguinte lista de núneros enteiros: **[1,5,9,10,12,20,35,40]**, utiliza a función intrínseca ***filter()*** para obter unha nova lista que conteña unicamente os que son divisibles por 5. Deberás usar unha **función lambda** na chamada a ***filter()***. Fai o mesmo pero agora creando a nova lista por comprensión e compara ambos resultados.

```python
lista = [1,5,9,10,12,20,35,40]
nova_lista = list(filter(lambda x: x%5 ==0, lista))
print(nova_lista)

# Mediante a creación dunha lista por comprensión:
nova_lista_2 = [x for x in lista if x%5 ==0]
print(nova_lista_2)
```

# Exercicio 4

Dada a seguinte lista de números: **[2.1, 20.9, 13.22, 15.8, 35.92]** utiliza a función intrínseca ***filter()*** para obter unha tupla que conteña unicamente os números da lista que estean comprendidos entre 10 e 20, ambolosdous incluídos. Deberás usar unha **función lambda** na chamada a ***filter()***.


```python
lista = [2.1, 20.9, 13.22, 15.8, 35.92]
nova_tupla = tuple(filter(lambda x:(x>=10 and x<=20), lista))
print(nova_tupla)

# Mediante a creación dunha lista por comprensión:
nova_tupla_2 = tuple([x for x in lista if (x>=10 and x<=20)])
print(nova_tupla_2)
```

# Exercicio 5

Dasa a seguinte lista de números: **[25,81,625,121]** utiliza a función intrínseca ***map()*** para crear outra lista que conteña as raíces cuadradas dos números da lista. Deberás usar unha **función lambda** na chamada a ***map()***.

```python
lista = [25,81,625,121]
nova_lista = list(map(lambda x:x**0.5,lista))
print(nova_lista)

# Mediante a creación dunha lista por comprensión:
nova_lista_2 = [x**0.5 for x in lista]
print(nova_lista_2)
```

# Exercicio 6

Dada a seguinte tupla de palabras: **('hola', 'que', 'tal', 'estás')** utiliza a función intrínseca ***map()*** para crear outra tupla que conteña as mesmas palabras pero escritas en maiúculas e en orden inverso **('ALOH', 'EUQ', 'LAT', 'SÁTSE')**. Deberás usar unha **función lambda** na chamada a ***map()***.


```python
palabras = ('hola', 'que', 'tal', 'estás')
nova_tupla = tuple(map(lambda x:x[::-1].upper(),palabras))
print(nova_tupla)

# Mediante a creación dunha lista por comprensión:
nova_tupla_2 = tuple([pal[::-1].upper() for pal in palabras])
print(nova_tupla_2)
```

# Exercicio 7

Dada a seguinte tupla de palabras **('hola', 'que', 'tal', 'estás')** utiliza a función intrínseca ***map()*** para crear outra tupla que conteña as mesmas palabras pero escritas de tal forma que se substitúan as vogais polo carácter asterisco **('h\*l\*', 'q\*\*', 't\*l', '\*st\*s')**. Deberá usar unha **función lambda** na chamada a ***map()***.

```python
def cambia_palabra(palabra):
    nova_palabra = ""
    for letra in palabra:
        if letra in "aeiouáéíóúAEIOUÁÉÍÓÚ":
            nova_palabra += "*"
        else:
            nova_palabra += letra
    return nova_palabra

palabras = ('hola', 'que', 'tal', 'estás')
nova_tupla = tuple(map(lambda x:"".join(["*" if letra in "aeiouáéíóúAEIOUÁÉÍÓÚ" else letra for letra in x]),palabras))
print(nova_tupla)

# Mediante a creación dunha lista por comprensión usando unha función definida con nome:
nova_tupla_2 = tuple([cambia_palabra(pal) for pal in palabras])
print(nova_tupla_2)

# Mediante a creación dunha lista por comprensión usando unha función lambda:
nova_tupla_3 = tuple([(lambda x:"".join(["*" if letra in "aeiouáéíóúAEIOUÁÉÍÓÚ" else letra for letra in x]))(pal) for pal in palabras])
print(nova_tupla_3)
```

# Exercicio 8

Dada a seguinte lista na que cada elemento é unha tupla con 3 valores: **datos = [(3, 7, 1), (11, 2, 25), (4, 12, 4), (7, 9, 2)]**, utiliza a función intrínseca **sorted()** para ordenar a mesma polo terceiro elemento de cada tupla. Deberás usar unha **función lambda** na chamada a ***sorted()***.


```python
datos = [(3, 7, 1), (11, 2, 25), (4, 12, 4), (7, 9, 2)]

datos_ordenados = sorted(datos, key= lambda x:x[2])
print(datos_ordenados)
```

# Exercicio 9

Dada a seguinte lista de palabras **["rojo", "azul", "amarillo", "naranja", "gris", "negro", "blanco"]**, utiliza o método ***sort()*** para ordenar a mesma pola lonxitude das mesmas. Deberás usar unha **función lambda** na chamada a ***sort()***.


```python
palabras = ["rojo", "azul", "amarillo", "naranja", "gris", "negro", "blanco"]

palabras.sort()
print(palabras)
palabras.sort(key= lambda x:len(x))
print(palabras)

# Mediante o uso da función intrínseca sorted()
palabras_ordenadas = sorted(palabras)
print(palabras_ordenadas)
palabras_ordenadas = sorted(palabras, key= lambda x:len(x))
print(palabras_ordenadas)
```

# Exercicio 10

Dada a seguinte lista de palabras **["rojo", "azul", "amarillo", "naranja", "gris", "negro", "blanco"]**, utiliza o método ***sort()*** para ordenar a mesma polo número de vogais que conteña cada palabra. Deberás usar unha **función lambda** na chamada a ***sort()***.


```python
def conta_vogais(palabra):
    conta = 0
    for letra in palabra:
        if letra in "aeiouáéíóúAEIOUÁÉÍÓÚ":
            conta +=1
    return conta


palabras = ["rojo", "azul", "amarillo", "naranja", "gris", "negro", "blanco"]

palabras.sort(key= lambda x:len([letra for letra in x if letra in "aeiouáéíóúAEIOUÁÉÍÓÚ"]))
print(palabras)

# Usando unha función con nome
palabras = ["rojo", "azul", "amarillo", "naranja", "gris", "negro", "blanco"]

palabras.sort(key=conta_vogais)
print(palabras)

# Mediante o uso da función intrínseca sorted()
palabras_ordenadas = sorted(palabras, key= lambda x:len([letra for letra in x if letra in "aeiouáéíóúAEIOUÁÉÍÓÚ"]))
print(palabras_ordenadas)
```
