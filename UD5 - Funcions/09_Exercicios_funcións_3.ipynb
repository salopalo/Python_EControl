# Exercicio 1

Crea unha función chamada **calcula_circunferencia** que reciba como argumentos:

- O valor do radio

E que devolva os seguintes valores:

- A lonxitude da circunferencia
- A área da circunferencia

Fai unha chamada á función usando como exemplo un valor de radio igual a 4, e comproba que devolve os valores correctos.

```python
import math

def calcula_circunferencia(radio):
    return 2*math.pi*radio, math.pi*radio**2

perimetro, area = calcula_circunferencia(5)

print(perimetro, area)
```

E que devolva os seguintes valores:

- O perímetro do rectángulo.
- A área do rectángulo.

# Exercicio 3

Crea unha función chamada **mezcla_mayus_minus** que reciba como argumento:

- Un string que conteña unha palabra.

E que devolva os seguintes valores:

- Outra palabra na que as posicións pares das letras estean en maiúsculas e as impares en minúsculas.
- Outra palabra na que as posicións pares das letras estean en minúsculas e as impares en maiúsculas.

Fai unha chamada á función usando como exemplo varias palabras. Para a palabra "exemplo": ExEmPlO, eXeMpLo.

```python
def mezcla_mayus_minus(palabra):
    palabra2 =""
    palabra3 =""
    for i in range(len(palabra)):
        if i %2:            
            palabra2 += palabra[i].upper()
            palabra3 += palabra[i].lower()
        else:
            palabra3 += palabra[i].upper()
            palabra2 += palabra[i].lower()
    return palabra2, palabra3

p1,p2 = mezcla_mayus_minus("exemplo")

print(p1,p2)


def mezcla_mayus_minus(palabra):
    palabra2 = ""
    palabra3 = ""
    for i, letra in enumerate(palabra): # enumerate devuelve una tupla con el numero de orden del iterable y su valor (ej. (0,"e") (1,"x") .. etc)
        if i % 2:  # Si el índice es impar
            palabra2 += letra.upper()
            palabra3 += letra.lower()
        else:  # Si el índice es par
            palabra2 += letra.lower()
            palabra3 += letra.upper()
            
    return palabra2, palabra3

p1,p2 = mezcla_mayus_minus("exemplo")

print(p1,p2)
```

- Outra palabra na que se substitúan as vogais por asteriscos(*) se a palabra remata en vogal, ou outra palabra na que se substitúan as vogais por almohadillas(#) se a palabra non remata en vogal.
- Un nº enteiro no que se contabilicen o nº de vogais da palabra.

Fai varias chamadas á función con distintas palabras para probar o seu funcionamento. Para a palabra "exemplo": \*x\*mpl\*, 3, para a palabra "monitores": m#n#t#r#s, 4. 

```python
def cambia_vocales(palabra):
    palabra2 = ""
    conta = 0
    for i in range(len(palabra)):
        if palabra[-1].lower() in ("a", "e", "i", "o", "u", "á", "é", "í", "ó", "ú"):
            if palabra[i].lower() in ("a", "e", "i", "o", "u", "á", "é", "í", "ó", "ú"):
                palabra2 += "*"
                conta +=1
            else: 
                palabra2 += palabra[i]
        else:
            if palabra[i].lower() in ("a", "e", "i", "o", "u", "á", "é", "í", "ó", "ú"):
                palabra2 += "#"
                conta +=1
            else: 
                palabra2 += palabra[i]
    return palabra2, conta

pal, cont = cambia_vocales("exemplo")
print(pal, cont)

pal, cont = cambia_vocales("monitores")
print(pal, cont)






def cambia_vocales(palabra):
    palabra2 = ""
    conta = 0
    ultima_es_vocal = palabra[-1].lower() in ("aeiouáéíóú")
    
    for letra in palabra:
        if letra.lower() in ("aeiouáéíóú"):
            palabra2 += "*" if ultima_es_vocal else "#"
            conta += 1
        else:
            palabra2 += letra
    
    return palabra2, conta
                
                
pal, cont = cambia_vocales("exemplo")
print(pal, cont)

pal, cont = cambia_vocales("monitores")
print(pal, cont)
            
        
```
