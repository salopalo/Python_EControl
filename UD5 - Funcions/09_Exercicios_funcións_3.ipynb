# Exercicios Funcións 3 (Devolución de varios valores)


# Exercicio 1

Crea unha función chamada **calcula_circunferencia** que reciba como argumentos:

- O valor do radio

E que devolva os seguintes valores:

- A lonxitude da circunferencia
- A área da circunferencia

Fai unha chamada á función usando como exemplo un valor de radio igual a 4, e comproba que devolve os valores correctos.

# Exercicio 2

Crea unha función chamada **calcula_rectángulo** que reciba como argumentos:

- O valor da base.
- O valor da altura.

E que devolva os seguintes valores:

- O perímetro do rectángulo.
- A área do rectángulo.

Fai unha chamada á función usando como exemplo os valores 3 e 5.

# Exercicio 3

Crea unha función chamada **mezcla_mayus_minus** que reciba como argumento:

- Un string que conteña unha palabra.

E que devolva os seguintes valores:

- Outra palabra na que as posicións pares das letras estean en maiúsculas e as impares en minúsculas.
- Outra palabra na que as posicións pares das letras estean en minúsculas e as impares en maiúsculas.

Fai unha chamada á función usando como exemplo varias palabras. Para a palabra "exemplo": ExEmPlO, eXeMpLo.

# Exercicio 4

Crea unha función chamada **cambia_vocales** que reciba como argumento:

- Un string que conteña unha palabra.

E que devolva os seguintes valores:

- Outra palabra na que se substitúan as vogais por asteriscos(*) se a palabra remata en vogal, ou outra palabra na que se substitúan as vogais por almohadillas(#) se a palabra non remata en vogal.
- Un nº enteiro no que se contabilicen o nº de vogais da palabra.

Fai varias chamadas á función con distintas palabras para probar o seu funcionamento. Para a palabra "exemplo": \*x\*mpl\*, 3, para a palabra "monitores": m#n#t#r#s, 4. 

# Exercicio 5

Crea unha función chamada **conta_máximos** que reciba como argumento:

- Unha lista de palabras (strings).

E que devolva os seguintes valores:

- Un nº enteiro indicando a lonxitude da palabra máis longa.
- Un nº enteiro indicando o nº vogais contido pola palabra que máis vogais conteña.
- Un nº enteiro indicando o nº de palabras total que conteña a lista.

Fai varias chamadas á función con distintas palabras para probar o seu funcionamento.
