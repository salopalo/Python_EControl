# Exercicio 1 (Uso de *args)

Crea unha función chamada **num_vectores_con_módulo_maior_de** que reciba un nº variable de vectores (coordenadas x,y) e conte o número deles cuxo módulo é maior que outro valor que se lle pase tamén como argumento.

```python
def num_vectores_con_modulo_maior_de(valor, *vectores):
    conta = 0
    for v in vectores:
        x,y = v
        if (x**2 + y**2)**0.5 > valor:
            conta +=1
    return conta


l = [(2,8), (12,3), (1,2)]

print(
    num_vectores_con_modulo_maior_de(5, (3,2), (2,7)),
    num_vectores_con_modulo_maior_de(5, *l))

```

```python
### SOLUCION


def num_vectores_con_módulo_maior_de(valor, *vectores):
    conta=0
    for V in vectores:
        módulo = (V[0]**2+V[1]**2)**0.5
        if módulo > valor:
            conta += 1
    return(conta)

print(num_vectores_con_módulo_maior_de(3,(1,2),(9,10),(-1,-1),(7,-3)))
lista_vectores=[(2,-5),(-8,9),(0,8),(-1,9)]
print(num_vectores_con_módulo_maior_de(7.9,*lista_vectores))

```

# Exercicio 2 (Uso de *kwargs)

Crea unha función chamada **planetas_maiores_de** que reciba un nº variable de nomes de planetas e o seu radio expresado en km. Debe devolver unha lista cos nomes dos planetas cuxo radio sexa menor que outro valor que se lle pase como argumento.

```python
def planetas_maiores_de(radio, **planetas):
    # lista_planetas = []
    # for p,r in planetas.items():
    #     if r < radio:
    #         lista_planetas.append(p)
    # return lista_planetas
    lista_planetas = [p for p,r in planetas.items() if r < radio]
    return lista_planetas



planetas={"Jupiter":69911, "Saturno":28232, "Urano":25362, "Neptuno":24622, "Tierra":6371, "Venus":6052, "Marte":3390, "Mercurio":2440}

print(planetas_maiores_de(15000, **planetas), "\n",
      planetas_maiores_de(15000, marte=25000, tierra=12500, saturno=45000, venus= 10200))



```

```python
### SOLUCION

def planetas_maiores_de(valor, **planetas):
   lista_planetas=[]
   for P,R in planetas.items():
      if R > valor:
         lista_planetas.append(P)
   return(lista_planetas)

planetas={"Jupiter":69911, "Saturno":28232, "Urano":25362, "Neptuno":24622, "Tierra":6371, "Venus":6052, "Marte":3390, "Mercurio":2440}
print(planetas_maiores_de(10000, **planetas))


```
