## Exercicio 1

Realiza un programa que pida desde teclado os seguintes datos relativos a un contribuínte:

- Soldo bruto anual
- Tipo de IRPF ao que está suxeito
- Idade

A partir de estes datos calcularase o importe que lle corresponde como pago do IRPF a Facenda, en base ao seguinte:

O cálculo do imposto farase aplicando a porcentaxe do tipo IRPF sobre o soldo bruto, e sumando despois 100 euros máis a todos os contribuíntes, a excepcion dos maiores de 50 anos que no seu lugar recibirán unha bonificación de 170 euros na súa declaración.

Para finalizar mostrarase por pantalla unha mensaxe co importe total a pagar.

```python
soldo = float(input("Introduza o soldo: \n\n"))
irpf = float(input("Introduza o tipo de IRPF: \n\n"))
idade = int(input("Introduza a idade: \n\n"))

imposto = soldo * irpf/100
if idade <= 50:
    imposto += 100
else:
    imposto -= 170
    
print(f"O imposto a pagar é de : {imposto}€")



```

```python
## SOLUCION

soldo = float(input("Introduza o soldo: "))
tipo = float(input("Introduza o tipo: "))
idade = int(input("Introduza a idade: "))
imposto = (soldo * tipo) / 100
if (idade <= 50):
    imposto += 100
else:
    imposto -= 170
print(f"O importe total a pagar é {imposto:.2f} euros")
```

```python
## SOLUCION



def calcula_IRPF(soldo,tipo,idade):
    imposto = soldo * tipo / 100
    if (idade <= 50):
        imposto += 100
    else:
        imposto -= 170
    return imposto

soldos = [0] ; tipos =[0] ; idades = [0] ; impostos = [0]
for i in range(1,6):
    soldos.append(float(input(f"Introduza o soldo do contribuínte nº{i}: ")))
    tipos.append(float(input(f"Introduza o tipo do contribuínte nº{i}: ")))
    idades.append(int(input(f"Introduza a idade do contribuínte nº{i}: ")))
    impostos.append(calcula_IRPF(soldos[i],tipos[i],idades[i]))
    print(f"O importe total a pagar polo contribuínte nº{i} é {impostos[i]:.2f} euros")
print("FIN DE PROGRAMA")
print("===============")
```
