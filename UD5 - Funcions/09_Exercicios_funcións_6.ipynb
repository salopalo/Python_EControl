# Exercicio 3

Dada a seguinte lista de números enteiros: **[1,5,9,10,12,20,35,40]**, utiliza a función intrínseca ***filter()*** para obter unha nova lista que conteña unicamente os que son divisibles por 5. Deberás usar unha **función lambda** na chamada a ***filter()***. Fai o mesmo pero agora creando a nova lista por comprensión e compara ambos resultados.

```python
lista = [1,5,9,10,12,20,35,40]

lista2 = list(filter(lambda a: a%5==0, lista))
print(lista2)
print(lista)

lista3 = [i for i in lista if i%5==0]
print(lista3)
```

# Exercicio 4

Dada a seguinte lista de números: **[2.1, 20.9, 13.22, 15.8, 35.92]** utiliza a función intrínseca ***filter()*** para obter unha tupla que conteña unicamente os números da lista que estean comprendidos entre 10 e 20, ambolosdous incluídos. Deberás usar unha **función lambda** na chamada a ***filter()***.


```python
lista = [2.1, 20.9, 13.22, 15.8, 35.92]

tupla = tuple(filter(lambda x: 10<=x<=20, lista))
print(tupla)
```

# Exercicio 5

Dasa a seguinte lista de números: **[25,81,625,121]** utiliza a función intrínseca ***map()*** para crear outra lista que conteña as raíces cuadradas dos números da lista. Deberás usar unha **función lambda** na chamada a ***map()***.

```python
import math

lista = [25,81,625,121]
list2 = list(map(lambda x: math.sqrt(x), lista))
print(list2)

lista = [25,81,625,121]
list2 = list(map(lambda x: x**(1/2), lista))
print(list2)
```

# Exercicio 6

Dada a seguinte tupla de palabras: **('hola', 'que', 'tal', 'estás')** utiliza a función intrínseca ***map()*** para crear outra tupla que conteña as mesmas palabras pero escritas en maiúculas e en orden inverso **('ALOH', 'EUQ', 'LAT', 'SÁTSE')**. Deberás usar unha **función lambda** na chamada a ***map()***.


```python
tupla = ('hola', 'que', 'tal', 'estás')

tupla2 = tuple(map(lambda x: x[::-1].upper(), tupla))
print(tupla2)
               
```

# Exercicio 7

Dada a seguinte tupla de palabras **('hola', 'que', 'tal', 'estás')** utiliza a función intrínseca ***map()*** para crear outra tupla que conteña as mesmas palabras pero escritas de tal forma que se substitúan as vogais polo carácter asterisco **('h\*l\*', 'q\*\*', 't\*l', '\*st\*s')**. Deberá usar unha **función lambda** na chamada a ***map()***.

```python
tupla = ('hola', 'que', 'tal', 'estás')
vogais = "aáeéiíoóuú"

tupla2 = tuple(map(lambda x: ''.join("*" if c.lower() in vogais else c for c in x),tupla))
print(tupla2)
```

# Exercicio 8

Dada a seguinte lista na que cada elemento é unha tupla con 3 valores: **datos = [(3, 7, 1), (11, 2, 25), (4, 12, 4), (7, 9, 2)]**, utiliza a función intrínseca **sorted()** para ordenar a mesma polo terceiro elemento de cada tupla. Deberás usar unha **función lambda** na chamada a ***sorted()***.


```python
datos = [(3, 7, 1), (11, 2, 25), (4, 12, 4), (7, 9, 2)]

ordenado = list(sorted(datos, key=lambda x: x[2]))
print(ordenado)

ordenado = list(sorted(datos, key=lambda x: x[2], reverse=True))
print(ordenado)
```

# Exercicio 9

Dada a seguinte lista de palabras **["rojo", "azul", "amarillo", "naranja", "gris", "negro", "blanco"]**, utiliza o método ***sort()*** para ordenar a mesma pola lonxitude das mesmas. Deberás usar unha **función lambda** na chamada a ***sort()***.


```python
lista = ["rojo", "azul", "amarillo", "naranja", "gris", "negro", "blanco"]

lista.sort(key=lambda x: len(x))
print(lista)

lista.sort()
print(lista)


ordenado = list(sorted(lista,key=lambda x: len(x)))
print(ordenado)

```

# Exercicio 10

Dada a seguinte lista de palabras **["rojo", "azul", "amarillo", "naranja", "gris", "negro", "blanco"]**, utiliza o método ***sort()*** para ordenar a mesma polo número de vogais que conteña cada palabra. Deberás usar unha **función lambda** na chamada a ***sort()***.


```python
lista = ["rojo", "azul", "amarillo", "naranja", "gris", "negro", "blanco"]
vogais = "aáeéiíoóuú"

lista.sort(key=lambda x: len([i for i in x if i.lower() in vogais]))
print(lista)

lista2 = sorted(lista, key=lambda x: len([i for i in x if i.lower() in vogais]))
print(lista2)
```
