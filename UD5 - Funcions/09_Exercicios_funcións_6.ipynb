# Exercicio 1

Crea unha **función lambda** que reciba tres argumentos e realice a suma dos dous primeiros e divida o resultado obtido entre o terceiro argumento. A función devolverá o resultado final da operación. Asigna dita función a unha variable chamada **calcular** e fai unha chamada á mesma usando distintos exemplos e comprobando que devolve os resultados esperados.

```python
calcular=lambda x,y,z: (x+y)/z
print(calcular(1,2,3))

(lambda x,y,z: (x+y)/z) (4,5,6)
```
