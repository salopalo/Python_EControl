# Exercicio 1

Crea unha función chamada **area_círculo** que reciba como argumento:

- O valor do radio en cm.

E que devolva o valor da área do círculo en cm2.

O cálculo da área farase elevando o radio ao cuadrado e multiplicando por pi (3.141592).

Fai unha chamada a dita función usando como exemplo o valor 7 e gardando o resultado na variable **área**, para escribir a continuación o seu valor na pantalla mediante un *print()*.


```python
import math 

def area_circulo(radio):
    return math.pi * radio**2

area = area_circulo(7)
print(area)
```

# Exercicio 2

Crea unha función chamada **comproba_parellas** que reciba como argumentos:

- Dous números enteiros.

E que devolva un string co valor "PARES" se ambolosdous números son pares, "IMPARES" se ambolosdous números son impares ou "PAR-IMPAR" se un deles e par e o outro impar.

Fai varias chamadas a dita función usando como exemplos os valores 3 e 5, 4 e 10, e 1 e 8, para escribir os resultados en pantalla e comprobar que devolve distintos resultados nas tres chamadas.


```python
def comproba_parellas(valor,valor2):
    if not valor%2 and not valor2%2:
        print("PARES")
    elif (not valor%2 and valor2%2) or (valor%2 and not valor2%2):
        print("PAR-IMPAR")
    else:
        print("IMPARES")
        
        
# print(4%2)
comproba_parellas(3,5)
comproba_parellas(4,10)
comproba_parellas(1,8)
comproba_parellas(1,1)
```

Crea unha función chamada **está_en_las_dos** que reciba como argumentos:

E que devolva o valor True se o número está nas dúas listas ou False se non está en dingunha ou está en só unha das listas.

# Exercicio 5

Crea unha función chamada **donde_está** que reciba como argumentos:

- Dúas listas de números enteiros.
- Un número enteiro para buscar nas listas.

E que devolva unha tupla co valor do indice no se que atopa en ambas listas. Se non aparece nas dúas ou só nunha delas debe devover o string "No está".  

Fai varias chamadas a dita función usando distintos exemplos.

```python
def donde_esta(lista, lista2, valor):
    res = []
    if valor in lista:
        res.append(lista.index(valor))
    else:
            res.append("No está.")
    if valor in lista2:
        res.append(lista2.index(valor))
    else:
        res.append("No está.")
    return tuple(res)
                
lista = [1,2,3,4,5]
lista2 = [3,5,6,7,0]

print(
donde_esta(lista, lista2, 5)
# donde_esta(lista, lista2, 0)
)


###########################################################################

def donde_esta2(valor, *listas):
    return tuple(lst.index(valor) if valor in lst else "No está." for lst in listas)

lista = [1,2,3,4,5]
lista2 = [3,5,6,7,0]

print(
donde_esta2(4, lista, lista2)
# donde_esta(lista, lista2, 0)
)

###########################################################################

#Si quieres evitar el doble recorrido (.index() ya busca internamente)

def donde_esta3(valor, *listas):
    def buscar(lst):
        try:
            return lst.index(valor)
        except ValueError:
            return "No está."
    
    return tuple(buscar(lst) for lst in listas)

lista = [1,2,3,4,5]
lista2 = [3,5,6,7,0]

print(
donde_esta3(3, lista, lista2)
# donde_esta(lista, lista2, 0)
)
    
```

# Exercicio 6

Crea unha función chamada **iniciales** que reciba como argumentos:

- Unha lista de nomes.
- Un string dunha letra.

E que devolva o número de elementos da lista que comezan por esa letra.  

Fai varias chamadas a dita función usando distintos exemplos.

```python
def iniciales(lista, caracter):
    conta = 0
    for i in lista:
        if i.upper().startswith(caracter.upper()):
            conta +=1
    return conta

lis = ["roger", "Novak", "Rafa", "Roberto", "Carlos"]
print(iniciales(lis, "R"))

##################################################################

def iniciales2(lista, caracter):
    return sum(1 for i in lista if i.upper().startswith(caracter.upper()))


lis = ["roger", "Novak", "Rafa", "Roberto", "Carlos"]
print(iniciales2(lis, "R"))
```

# Exercicio 7

Crea unha función chamada **sumar_letras** que reciba como argumentos:

- Unha lista de nomes.
- Un string dunha letra.

E que devolva o número de veces que se repite esa letra sumando todos os nomes da lista.  

Fai varias chamadas a dita función usando distintos exemplos.

```python
def sumar_letras(lista, caracter):
    conta = 0
    for i in lista:
        conta += i.upper().count(caracter.upper())
    return conta

lis = ["roger", "Novak", "Rafa", "Roberto", "Carlos"]
print(sumar_letras(lis, "a"))



def sumar_letras2(lista, caracter):
    return sum(i.upper().count(caracter.upper()) for i in lista)

lis = ["roger", "Novak", "Rafa", "Roberto", "Carlos"]
print(sumar_letras2(lis, "a"))
```
