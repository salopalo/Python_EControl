## Exercicio 1

Dada a seguinte lista na que se indica a temperatura media de cada mes na cidade de Vigo:

 **temperaturas_Vigo = [8.8, 9.6, 11.7, 12.8, 15.3, 17.9, 19.8, 20.2, 18.5, 15.3, 11.5, 9.4]** 

Calcula mediante ***"slicing"*** e máis as funcións intrínsecas **sum** e **len** a media de cada trimestre, semestre e a media anual e mostra o resultado por pantalla.

```python
temperaturas_Vigo = [8.8, 9.6, 11.7, 12.8, 15.3, 17.9, 19.8, 20.2, 18.5, 15.3, 11.5, 9.4]
media_trim1 = sum(temperaturas_Vigo[0:3])/len(temperaturas_Vigo[0:3])
media_trim2 = sum(temperaturas_Vigo[3:6])/len(temperaturas_Vigo[3:6])
media_trim3 = sum(temperaturas_Vigo[6:9])/len(temperaturas_Vigo[6:9])
media_trim4 = sum(temperaturas_Vigo[9:12])/len(temperaturas_Vigo[9:12])
media_sem1 = sum(temperaturas_Vigo[0:6])/len(temperaturas_Vigo[0:6])
media_sem2 = sum(temperaturas_Vigo[6:12])/len(temperaturas_Vigo[6:12])
media = sum(temperaturas_Vigo) / len(temperaturas_Vigo)
print(f"A media do primeiro trimestre é: {media_trim1:.1f}")
print(f"A media do segundo trimestre é: {media_trim2:.1f}")
print(f"A media do terceiro trimestre é: {media_trim3:.1f}")
print(f"A media do cuarto trimestre é: {media_trim4:.1f}")
print(f"A media do primeiro semestre é: {media_sem1:.1f}")
print(f"A media do segundo semestre é: {media_sem2:.1f}")
print(f"A media anual é: {media:.1f}")
```

## Exercicio 3

A partir da listas seguintes, que conteñen o nome de varias cidades galegas e o nº de habitantes das mesmas:

 **cidades = ["Santiago de Compostela", "Vigo", "A Coruña", "Ourense", "Lugo", "Pontevedra"]**
 
 **habitantes = [ 98179, 292374, 244700, 103756, 97211, 82828]**

Introduce ao final de cada lista o dato correspondente a Ferrol (64158 habitantes) co método correspondente. A continuación ordena a lista de habitantes de maior a menor poboación. Fai que a lista de cidades se corresponda en orden co número de habitantes da outra lista (usa un método para borrar a cidade de "Santiago de Compostela" e despois insértala na lista na posición correcta).

Crea un diccionario por comprensión chamado **poboación** no que as claves sexan os nomes das cidades e os valores sexan o dato correspondente á poboación de cada cidade.

Para crear o diccionario por comprensión utiliza as dúas listas: **cidades** e **habitantes** e un **"for"** que itere desde 0 ata 6.

```python
cidades = ["Santiago de Compostela", "Vigo", "A Coruña", "Ourense", "Lugo", "Pontevedra"]
habitantes = [ 98179, 292374, 244700, 103756, 97211, 82828]
cidades.append('Ferrol')
habitantes.append(64158)
habitantes.sort(reverse=True)
cidades.remove("Santiago de Compostela")
cidades.insert(3,"Santiago de Compostela")
print(cidades)
print(habitantes)
poboación = {cidades[i] : habitantes[i] for i in range(7)}
print(poboación)
```

## Exercicio 4

Utilizando o seguinte dicionario que recolle o nº de días de chuvia anual de varias cidades europeas:

**días_chuvia = {'Copenhague': 251, 'Oslo': 232, 'Londres': 226, 'Bruselas': 220, 'Ámsterdam': 214, 'Helsinki': 203, 'Estocolmo': 164, 'Viena': 160, 'Praga': 143, 'Santiago de Compostela': 140, 'Dublín': 139}**

Mostra por pantalla o nome da cidade que teña 214 días de chuvia anual. Para atopala recorre o diccionario cun **"for"** ata dar co primeiro valor que coincida co dato.

Utilizando a función ***max()*** mostra por pantalla a cidade co maior nº de días de chuvia anual e o seu valor.

```python
días_chuvia = {'Copenhague': 251, 'Oslo': 232, 'Londres': 226, 'Bruselas': 220, 'Ámsterdam': 214, 'Helsinki': 203, \
               'Estocolmo': 164, 'Viena': 160, 'Praga': 143, 'Santiago de Compostela': 140, 'Dublín': 139}

for clave,valor in días_chuvia.items():
    if valor == 214:
        print(f"A cidade que ten un nº de días de chuvia anual igual a 214 é: {clave}")
        break

# maximo = max(días_chuvia.values()) Así solo compara los valores y devuelve el más alto, pero nos faltaría la ciudad (la clave)

posicion = max(días_chuvia,key=días_chuvia.get)  # Así compara los valores y devuelve la posición (la clave) del más alto
maximo = días_chuvia[posicion]                    # Indexando el diccionario obtenemos el valor 
print(f"A cidade con maior nº de días de chuvia anual é {posicion} cun valor igual a {maximo}") 
```
