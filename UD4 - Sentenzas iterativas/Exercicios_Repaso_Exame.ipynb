## Exercicio 1

Dada a seguinte lista na que se indica a temperatura media de cada mes na cidade de Vigo:

 **temperaturas_Vigo = [8.8, 9.6, 11.7, 12.8, 15.3, 17.9, 19.8, 20.2, 18.5, 15.3, 11.5, 9.4]** 

Calcula mediante ***"slicing"*** e máis as funcións intrínsecas **sum** e **len** a media de cada trimestre, semestre e a media anual e mostra o resultado por pantalla.

```python
temperaturas_Vigo = [8.8, 9.6, 11.7, 12.8, 15.3, 17.9, 19.8, 20.2, 18.5, 15.3, 11.5, 9.4]

media_T1 = round(sum(temperaturas_Vigo[:3])/len(temperaturas_Vigo[:3]),1)
media_T2 = round(sum(temperaturas_Vigo[3:6])/len(temperaturas_Vigo[:3]),1)
media_T3 = round(sum(temperaturas_Vigo[6:9])/len(temperaturas_Vigo[:3]),1)
media_T4 = round(sum(temperaturas_Vigo[9:])/len(temperaturas_Vigo[:3]),1)

media_S1 = round(sum(temperaturas_Vigo[:6])/len(temperaturas_Vigo[:6]),1)
media_S2 = round(sum(temperaturas_Vigo[6:])/len(temperaturas_Vigo[6:]),1)

media_anual = round(sum(temperaturas_Vigo)/len(temperaturas_Vigo),1)

print(f"Medias Trimestrais: T1:{media_T1}, T2:{media_T2}, T3:{media_T3}, T4:{media_T4}")
print(f"Medias Semestrais: S1:{media_S1}, S2:{media_S2}")
print(f"Media anual: {media_anual}")
```

## Exercicio 4

Utilizando o seguinte dicionario que recolle o nº de días de chuvia anual de varias cidades europeas:

**dias_chuvia = {'Copenhague': 251, 'Oslo': 232, 'Londres': 226, 'Bruselas': 220, 'Ámsterdam': 214, 'Helsinki': 203, 'Estocolmo': 164, 'Viena': 160, 'Praga': 143, 'Santiago de Compostela': 140, 'Dublín': 139}**

Mostra por pantalla o nome da cidade que teña 214 días de chuvia anual. Para atopala recorre o diccionario cun "for" ata dar co primeiro valor que coincida co dato.

Utilizando a función max() mostra por pantalla a cidade co maior nº de días de chuvia anual e o seu valor.


```python
dias_chuvia = {'Copenhague': 251, 'Oslo': 232, 'Londres': 226, 'Bruselas': 220, 'Ámsterdam': 214, 'Helsinki': 203, 'Estocolmo': 164, 'Viena': 160, 'Praga': 143, 'Santiago de Compostela': 140, 'Dublín': 139}

for i in dias_chuvia:
    if dias_chuvia[i] == 214:
        print(f"A primeira cidade con 214 dias de chuvia é: {i}.")
        break

cidade_max = max(dias_chuvia, key=dias_chuvia.get)
                  
print(f"A cidade co maior número de días de chuvia anual é {cidade_max}, cun valor de {dias_chuvia[cidade_max]}")

```
