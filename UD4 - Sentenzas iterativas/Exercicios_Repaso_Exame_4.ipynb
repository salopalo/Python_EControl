## Exercicio 1

A partir da listas seguintes, que conteñen o nome de varios produtos e o seu prezo:

 **produtos = ["Refresco cola", "Pizza", "Helado", "Patatas onduladas", "Salsa barbacoa"]**
 
 **prezos = [0.90, 3.29, 1.52, 1.91, 2.09]**

Crea un diccionario por comprensión chamado **inventario** no que as claves sexan os nomes dos produtos e os valores sexan o prezo correspondente ao produto. Para crear o diccionario por comprensión utiliza as dúas listas e un ZIP.

```python
produtos = ["Refresco cola", "Pizza", "Helado", "Patatas onduladas", "Salsa barbacoa"]
prezos = [0.90, 3.29, 1.52, 1.91, 2.09]

inventario = { nombre:precio for nombre, precio in zip(produtos,prezos)}

print(inventario)
```

## Exercicio 2

Utilizando o seguinte dicionario que é o resultado do exercicio anterior:

**inventario = {'Refresco cola': 0.9, 'Pizza': 3.29, 'Helado': 1.52, 'Patatas onduladas': 1.91, 'Salsa barbacoa': 2.09}**

Mostra por pantalla o nome do produto que teña un prezo igual a 1.91 euros. Para atopalo percorre o dicionario cun **"for"** ata dar co primeiro valor que coincida co dato. Atopa unha solución alternativa obtendo as listas de claves e de valores e usando o método index() da lista de valores.


```python
inventario = {'Refresco cola': 0.9, 'Pizza': 3.29, 'Helado': 1.52, 'Patatas onduladas': 1.91, 'Salsa barbacoa': 2.09}




for nom, prez in inventario.items():
    if prez == 1.91:
        print(f"O produto que vale 1.91€ é {nom}")
        
claves = list(inventario.keys())
valores = list(inventario.values())

print(f"O produto que vale 1.91€ é {claves[valores.index(1.91)]}")
        
```

## Exercicio 5

Utilizando o mesmo dicionario do exercicio anterior:

**inventario = {'Refresco cola': 0.9, 'Pizza': 3.29, 'Helado': 1.52, 'Patatas onduladas': 1.91, 'Salsa barbacoa': 2.09}**


Crea unha tupla de nome **produtos_baratos** que conteña os nomes dos produtos cun prezo menor de 2 euros.

Crea unha tupla de nome **produtos_caros** que conteña os nomes dos produtos cun prezo igual ou superior a 2 euros.

Cra un novo dicionario de nome **inventario_nomes_simples** que conteña os produtos (e o seu prezo) que non teñan un nome composto por dúas palabras.


```python
inventario = {'Refresco cola': 0.9, 'Pizza': 3.29, 'Helado': 1.52, 'Patatas onduladas': 1.91, 'Salsa barbacoa': 2.09}

produtos_baratos = tuple(nom for nom,prez in inventario.items() if prez < 2)
print(produtos_baratos)

produtos_caros = tuple(nom for nom,prez in inventario.items() if prez >=2)
print(produtos_caros)

inventario_nomes_simples = {nom:prez for nom, prez in inventario.items() if len(nom.split())==1}
print(inventario_nomes_simples)
```
