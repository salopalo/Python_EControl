# Erros e excepcións 

Os erros teñen tres sabores básicos:

- **Erros de sintaxe (syntax errors):** 
  - Ocorren cando o teu código non é válido (adoita ser fácil de amañar, por exemplo, se che falta o `:` ó final do bucle)
- **Erros en tempo de execución (Runtime errors):** 
  - Se o teu código non contén erros de sintaxe, aínda pode ser que falle ó executarse. Podería ser, por exemplo, un texto introducido pola persoa que o usa (ás veces é fácil de corrixir, por exemplo cando les datos incorrectos)
- **Erros semánticos (Semantic errors):** 
  - O teu código corre sen problema, pero o resultado no é o que esperabas (como Python non sabe que esperabas, este tipo de erros pode ser difícil de depurar)


Neste apartado ímonos fixar en que facer para xestionar limpamente os **erros en tempo de execución**.

```python
while True:
    try:
        filename = input("Por favor, teclea o nome dun ficheiro: ")
        if len(filename)<6:
            print('Os nomes de ficheiros teñen que ter como pouco 6 caracteres, proba outra vez...')
            continue

        # === Added: ===
        elif ' ' in filename:
            print('Os nomes de ficheiros non poden ter espazos, proba outra vez...')
            continue
            
        else:
            open(filename)
            # Read data here ...
            break
    
    except OSError as err:
        print(err)
        print('\n OLLO: Non dou lido o ficheiro. Usarei valores por defecto.')
        # Aquí faríase outra cousa ...
        break

    except: 
        # podes aproveitar para facer algo de limpeza antes de levantar o erro
        print('Erro inesperado:')
        raise
```
