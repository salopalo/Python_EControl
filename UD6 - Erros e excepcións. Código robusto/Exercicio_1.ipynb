Exercicio nº 1  
Dada a seguinte lista na que se indica a temperatura media rexistrada cada mes na
cidade de Vigo:  
temperaturas_Vigo = [8.8, 9.6, 11.7, 12.8, 15.3, 17.9, 19.8, 20.2, 18.5, 15.3, 11.5, 9.4 ]  
Realiza un programa Python no que se solicite un mes inicial e un mes final para calcular a media de temperaturas no período comprendido entre eses dous meses. O programa debe de ser robusto, de tal maneira que pida o mes inicial e o mes final para facer o cálculo tendo en conta o seguinte:  
• Os meses só se admiten se están escritos polo seu nome en galego.  
• O mes inicial debe ser cronoloxicamente anterior ao mes final.  
• A entrada de datos seguirá solicitándose mentres non se cumpran as condicións anteriores.

```python





temperaturas_Vigo = [8.8, 9.6, 11.7, 12.8, 15.3, 17.9, 19.8, 20.2, 18.5, 15.3, 11.5, 9.4 ]  
meses = ["xaneiro", "febreiro", "marzo", "abril", "maio", "xuño", "xullo", "agosto", "setembro", "outubro", "novembro", "decembro"]


while True:
    try:
        mes_inicial = input("Introduza o mes inicial: ")
        if not mes_inicial.lower() in meses:
            raise ValueError("Nome de mes non válido. Debe estar escrito en galego.")
        else:
            mes_inic_indice = meses.index(mes_inicial.lower())
        if mes_inic_indice == 11:
            raise ValueError("Nome de mes non válido. O primeiro mes non pode ser decembro.")
        while True:
            try:
                mes_final = input("Introduza o mes final: ")
                if not mes_final.lower() in meses:
                    raise ValueError("Nome de mes non válido. Debe estar escrito en galego.")
                else:
                    mes_fin_indice = meses.index(mes_final.lower())
                if mes_fin_indice <= mes_inic_indice:
                    raise ValueError(f"Nome de mes non válido. Debe de ser posterior ao mes inicial ({meses[mes_inic_indice]})")
            except ValueError as error:
                print(error)
            else: break
    except ValueError as error:
        print(error)
        
    else:
        media_temp = round(sum(temperaturas_Vigo[mes_inic_indice:mes_fin_indice+1]) / len(temperaturas_Vigo[mes_inic_indice:mes_fin_indice+1]),2)

        print(media_temp) 
        break
        
        
        # print(mes_inic_indice, mes_fin_indice)

    
#___________________________________ SOLUCION ________________________________________________

temperaturas_Vigo = [8.8, 9.6, 11.7, 12.8, 15.3, 17.9, 19.8, 20.2, 18.5, 15.3, 11.5, 9.4]
meses_galego = ('xaneiro','febreiro','marzo','abril','maio','xuño','xullo','agosto','setembro','outubro','novembro','decembro')

print("CÁLCULO DA MEDIA DE TEMPERATURAS ENTRE DOUS MESES DO ANO")
while True:
    try:
        mes_inicio = input("Introduza o mes inicial:")    
        if mes_inicio.lower() not in meses_galego:
            raise ValueError("Nome de mes non válido. Debe de estar escrito en galego.")
        if mes_inicio.lower() == 'decembro':
            raise ValueError("Nome de mes non válido. O primeiro mes non pode ser decembro.")
        while True:
            try:
                mes_fin = input("Introduza o mes final:")
                if mes_fin.lower() not in meses_galego:
                    raise ValueError("Nome de mes non válido. Debe de estar escrito en galego.")
                if meses_galego.index(mes_inicio.lower()) >= meses_galego.index(mes_fin.lower()):
                    raise ValueError(f"Nome de mes non válido. Debe de ser posterior ao mes inicial ({mes_inicio.lower()}).")
            except ValueError as error:
                print(error)
            else:
                break    
    except ValueError as error:
        print(error)
    else:
        inicio = meses_galego.index(mes_inicio.lower())
        fin = meses_galego.index(mes_fin.lower())
        media = sum(temperaturas_Vigo[inicio:fin+1])/len(temperaturas_Vigo[inicio:fin+1])
        print(f"A media de temperaturas de {mes_inicio.lower()} a {mes_fin.lower()} é: {media}")
        break
```
