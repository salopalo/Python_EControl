```python
########################################## solucion ############################################

class CidadeNonRexistrada(Exception):
    def __init__(self, nome_cidade, message="Non rexistrada en días_chuvia"):
        self.nome_cidade = nome_cidade
        self.message = message
        super().__init__(self.message)
    def __str__(self):
        return f"Cidade: {self.nome_cidade}. {self.message}"

días_chuvia = {'Copenhague': 251, 'Oslo': 232, 'Londres': 226, 'Bruselas': 226, 'Ámsterdam': 214, 'Helsinki': 203,\
               'Estocolmo': 164, 'Viena': 164, 'Praga': 164, 'Santiago de Compostela': 140, 'Dublín': 139}
días_chuvia_lower = {C.lower(): V for C, V in días_chuvia.items()}

while True:
    try:
        print("** CONSULTA DA CHUVIA ANUAL EN EUROPA **")
        print("1 - Consulta por cidade")
        print("2 - Consulta por nº de días de chuvia")
        print("3 - Finalizar programa")
        print("-------------------------------------")
        op = int(input("Opción ->"))
        
        if (op == 1):
            cidade = input("Introduza a cidade:")
            if cidade.lower() not in días_chuvia_lower.keys():
                raise CidadeNonRexistrada(cidade)
            else:
                print(f"O nº de días de chuvia anuais en {cidade} é de {días_chuvia_lower[cidade.lower()]}")
        elif (op == 2):
            n_días = int(input("Introduza o nº de días:"))
            if  n_días < 0:
                raise ValueError("Non se permite un nº de días negativo")
            elif n_días not in días_chuvia.values():
                raise ValueError("Non se atoparon cidades con ese número de días de chuvia")
            else:
                for C, V in días_chuvia.items():
                    if n_días == V:
                        print(C)          
        elif (op == 3):
            print("Fin de programa")
            break
        else:
            raise ValueError("Opción non válida (teclee 1, 2, ou 3)")
    
    except (ValueError,CidadeNonRexistrada) as error:
        print(error)

```

```python

inventario = {'Refresco cola': 0.9, 'Pizza': 3.29, 'Helado': 1.52, 'Patatas onduladas': 1.91, 'Salsa barbacoa': 2.09, 'Panecillos': 0.9}

class ProdutoNonRexistrado(Exception):
    def __init__(self, prod, mens="Non rexistrado no inventario."):
        self.mens = mens
        self.prod = prod
        super().__init__(self.mens)
        
    def __str__(self):
        return f"Produto: {self.prod}. {self.mens}"
        
while True:
    try:
        print("**CONSULTA DE PRODUTOS DE SUPERMERCADO**")
        print("1 - Consulta por nome de produto")
        print("2 - Consulta por prezo")
        print("3 - Finalizar programa")
        print("----------------------------------------")
        option = int(input("Introduza a opción: "))
        print(f"Opción -> {option}")
        
        if option == 1:
            try:
                nome = input("Introduza o nome do produto: ")
                print(f"Introduza o nome do produto: {nome}")
                if nome.lower() not in str(inventario.keys()).lower():
                    raise ProdutoNonRexistrado(nome)
                else:
                    for i in inventario:
                        if nome.lower() == i.lower():
                            print(f"O prezo do produto {nome.lower()} é de {inventario[i]} euros.")
                
            except ProdutoNonRexistrado as error:
                print(error)
              
        if option == 2:
            try:
                prezo = input("Introduza o prezo: ")
                print(f"Introduza o prezo: {prezo}\n")
                prezo = float(prezo)
            except ValueError as error:
                print("O prezo debe ser un valor numérico")
            else:
                try:
                    if prezo <= 0:
                        raise ValueError("Non se permite un prezo cero ou negativo")
                    elif prezo not in inventario.values():
                        raise ValueError ("Non se atoparon produtos con ese prezo")
                    else:
                        for i in inventario:
                            if prezo == inventario[i]:
                                print(i)    
                                
                except ValueError as e:
                    print(e)            
        
        if option == 3:
            print("Fin de programa")
            break
        
        else:
            if isinstance(option, int) and option not in (1,2,3):
                print("Opción non válida (teclee 1, 2 ou 3)")
            else:
                raise ValueError()       
        
    except (ValueError, TypeError) as error:
        print(error)
    
    
    
    
    
    
    

```

```python
################################  SOLUCION   ########################################3


class ProdutoNonRexistrado(Exception):
    def __init__(self, nome_produto, message="Non rexistrado no inventario."):
        self.nome_produto = nome_produto
        self.message = message
        super().__init__(self.message)
    def __str__(self):
        return f"Produto: {self.nome_produto}. {self.message}"

inventario = {'Refresco cola': 0.9, 'Pizza': 3.29, 'Helado': 1.52, 'Patatas onduladas': 1.91, 'Salsa barbacoa': 2.09, 'Panecillos':0.9}

inventario_lower = {P.lower(): V for P, V in inventario.items()}

while True:
    try:
        print("** CONSULTA DE PRODUTOS DE SUPERMERCADO **")
        print("1 - Consulta por nome de produto")
        print("2 - Consulta por prezo")
        print("3 - Finalizar programa")
        print("-------------------------------------")
        op = int(input("Opción ->"))
        
        if (op == 1):
            produto = input("Introduza o nome do produto:")
            if produto.lower() not in inventario_lower.keys():
                raise ProdutoNonRexistrado(produto)
            else:
                print(f"O prezo do produto {produto} é de {inventario_lower[produto.lower()]:.2f} euros")
        elif (op == 2):
            prezo_prod = float(input("Introduza o prezo:"))
            prezo_prod = round(prezo_prod,2)
            if  prezo_prod <= 0:
                raise ValueError("Non se permite un prezo cero ou negativo")
            elif prezo_prod not in inventario.values():
                raise ValueError("Non se atoparon produtos con ese prezo")
            else:
                for P, V in inventario.items():
                    if prezo_prod == V:
                        print(P)          
        elif (op == 3):
            print("Fin de programa")
            break
        else:
            raise ValueError("Opción non válida (teclee 1, 2, ou 3)")
    
    except (ValueError,ProdutoNonRexistrado) as error:
        print(error)

```

Incluir a xestión de erros, incluindo algunha excepción personalizada.

```python



class LibroNonAtopado(Exception):
    def __init__(self, titulo, mens="Este libro non se atopa rexistrado."):
        self.titulo = titulo
        self.mens = mens
        super().__init__(self.mens)
        
    def __str__(self):
        return f"{self.titulo} -> {self.mens}"
  

prestamos = {}  



while True:

    
    try:
        print("\n**XESTION DE BIBLIOTECA**")
        print("1 - Préstamo de libro")
        print("2 - Retorno de libro")
        print("3 - Consulta de libro")
        print("4 - Consulta de usuario")
        print("5 - Mostrar a colección de préstamos.")
        print("6 - Fin do programa")
        print("--------------------------")

        try:
            opt = int(input("Introduza a opción: "))
            print(f"Opcion ->{opt}")
        except ValueError as e:
            print("Opción non válida, debe introducir un número entre 1 e 6.")
            opt = None
        
        if opt == 1:
            lib = input("Introduza o titulo do libro.")
            user = input("Introduza o nome do usuario.")
            atopado = False
            for us in prestamos:
                if user.lower() == us.lower():
                    prestamos[us].append(lib)
                    atopado = True
            if not atopado:
                prestamos[user] = [lib]
            print(f"O libro {lib} prestouse ao usuario {user}")
                    
        elif opt == 2:
            lib = input("Introduza o titulo do libro.")
            user = input("Introduza o nome do usuario.")
            atopado_us = False
            atopado_lib = False
            for us, li in prestamos.items():
                if user.lower() == us.lower():
                    atopado_us = True
                    for libro in li:
                        if lib.lower() == libro.lower():
                            prestamos[us].remove(libro)
                            print(f"Retornado o libro {lib} do usuario {user}.")
                            atopado_lib = True
            if not atopado_us:
                print(f"O usuario {user} non ten préstamos.")
            elif not atopado_lib:
                print(f"O usuario {user} non ten a préstamo o libro {lib}. Non se pode proceder á devolución.")
            
        elif opt == 3:
            lib = input("Introduza o titulo do libro: ")
            atopado = False
            for us, li in prestamos.items():
                for libro in li:
                    if lib.lower() == libro.lower():
                        print(f"O libro {lib} prestouse ao usuario {us}.")
                        atopado = True 
            if not atopado:
                raise LibroNonAtopado(lib)
    
        elif opt == 4:
            user = input("Introduza o nome do usuario: ")
            atopado = False
            for us, li in prestamos.items():
                if user.lower() == us.lower():
                    libros = ", ".join(li) if li else "ningun libro"
                    print(f"O usuario {user} ten en préstamo os libros: {libros}")
                    atopado = True 
            if not atopado:
                print("O usuario non existe.")
                
        elif opt == 5:
            if prestamos:
                for us, li in prestamos.items():
                    libros = ", ".join(li) if li else "ningun libro"
                    print(f"Usuario: {us}. Libros a préstamo: {libros}")
            else:
                print("Base de datos vacía.")
                     
        elif opt == 6:
            print("Fin do programa")
            break
        
        else:
            if isinstance(opt, (int, float)) and opt not in (1,2,3,4,5,6):
                raise ValueError ("Opción non válida, teclee unha opción entre 1 e 6.")
            

    except (ValueError, TypeError, LibroNonAtopado) as error:
        print (error)
        
        
    
    
    
```
