### 1.- Convierte de Farenheit a Celsius añadiendo un control de errores en la conversion a float de la temperatura introducida

```python
print ("Conversion de ºF a ºC")
while True:
    try:
        farenheit = float(input("Introduza a temperatura en ºF:"))
    except (ValueError):
        print("O tipo de dato introducido non é válido. Intenteo de novo")
    else:
        celsius = round((farenheit-32)/1.8,2)

        print(f"A temperatura en ºC sería: {celsius}ºC")
        break

```

### 2.- Dado o seguinte dicionario que recolle o nº de días de chuvia anual de varias cidades europeas:

días_chuvia = {'Copenhague': 251, 'Oslo': 232, 'Londres': 226, 'Bruselas': 226, 'Ámsterdam': 214, 'Helsinki': 203, 'Estocolmo': 164, 'Viena': 164, 'Praga': 164, 'Santiago de Compostela': 140, 'Dublín': 139}  

Realiza un programa que mostre un menú con dúas opcións para consultar o contido do mesmo: consultar por nome de cidade ou consultar por nº de días de chuvia. O programa debe ser robusto de tal maneira que teña en conta o seguinte:  
• Se consultamos por cidade, debe de contemplarse o caso de que a mesma non estea no dicionario. Nese caso lanzaremos unha excepción personalizada de nome CidadeNonRexistrada na que se construirá unha mexase coma a do exemplo:

• No caso da consulta por cidade, debe de admitirse o nome de cidade independentemente da súa escritura en maiúsculas, minúsculas ou unha mestura de ambas, como podes ver no exemplo:  

![](Ex_2_2.jpg)

• Se consultamos por nº de días de chuvia, debe de controlarse que se introduza un enteiro non negativo. Nese caso podemos usar a excepción ValueError enviándolle unha mensaxe como a que podes ver no exemplo:  

![](Ex_2_3.jpg)

• Na consulta por nº de días de chuvia, tamén debe de controlarse que o número buscado estea no dicionario. En caso contrario, lanzaremos a excepción ValueError enviándolle unha mensaxe como a do exemplo:  

![](Ex_2_4.jpg)

• No caso de que varias cidades teñan o mesmo nº de días de chuvia, deberán de mostrarse todas elas, como podes ver no exemplo:

![](Ex_2_5.jpg)

• Na introdución da opción do menú, debe de controlarse que se trate dun número e ademais que sexa 1, 2 ou 3.

![](Ex_2_6.jpg)

• Nos casos de introdución de datos que non sexan números enteiros, tanto no menú inicial como cando se pide o nº de días de chuvia trataremos a excepción correspondente:  

![](Ex_2_7.jpg)

• O programa finalizará ao introducir a opción 3.

![](Ex_2_8.jpg)




```python


class CidadeNonRexistrada(Exception):
    def __init__(self, cidade, message=f"Non rexistrada en dias_chuvia"):
        self.cidade = cidade
        self.message = message
        super().__init__(self.message)
        
    def __str__(self):
        return f"Cidade: {self.cidade}. {self.message}"
    
    
        

dias_chuvia = {'Copenhague': 251, 'Oslo': 232, 'Londres': 226, 'Bruselas': 226, 'Ámsterdam': 214, 'Helsinki': 203, 'Estocolmo': 164, 'Viena': 164, 'Praga': 164, 'Santiago de Compostela': 140, 'Dublín': 139} 

while True:
    try: 
        print("\n** CONSULTA DA CHUVIA ANUAL EN EUROPA **\n"
            "1 - Consulta por cidade\n"
            "2 - Consulta por nº de días de chuvia\n"
            "3 - Finalizar programa\n"
            "----------------------------------------"
            )
        option = int(input("Opcion -> "))
        
        if option == 1:
            print(f"Opcion -> {option}")
            try: 
                cidade = input("Introduza a cidade: ")
                print(f"Introduza a cidade: {cidade}")
                encontrado = 0
                for cid, val in dias_chuvia.items():
                    if cidade.lower() == cid.lower():
                        encontrado = 1
                        break
                if encontrado:
                    print(f"O nº de dias de chuvia anuais en {cid} é de {val} ")
                else:
                    raise CidadeNonRexistrada(cidade)
            except CidadeNonRexistrada as e:
                print(f"Erro: {e}")
        
        elif option == 2:
            print(f"Opcion -> {option}")
            try:
                dias = int(input("Introduza o nº de días: "))
                print(f"Introduza o nº de días: {dias}")
                if dias < 0:
                    raise ValueError("Non se permite un numero de días negativo")
                elif dias not in dias_chuvia.values():
                    raise ValueError("Non se atoparon cidades con ese número de días de chuvia.")
                else:
                    for c,d in dias_chuvia.items():
                        if dias == d:
                            print(c)
            
            except ValueError as e:
                print(e)
                
                
        elif option == 3:
            print("Fin del programa")
            break
        
        else:
            print(f"Opcion -> {option}")
            if isinstance(option, (int)):
                raise ValueError("Opción non válida (teclee 1, 2 ou 3)") 
            elif not isinstance(option, (int)):
                raise TypeError()
           
    except (ValueError) as error:
        print(error)   




```
