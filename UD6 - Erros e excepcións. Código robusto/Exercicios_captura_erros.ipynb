### 1.- Convierte de Farenheit a Celsius añadiendo un control de errores en la conversion a float de la temperatura introducida

```python
print ("Conversion de ºF a ºC")
while True:
    try:
        farenheit = float(input("Introduza a temperatura en ºF:"))
    except (ValueError):
        print("O tipo de dato introducido non é válido. Intenteo de novo")
    else:
        celsius = round((farenheit-32)/1.8,2)

        print(f"A temperatura en ºC sería: {celsius}ºC")
        break

```

```python
########################################## solucion ############################################

class CidadeNonRexistrada(Exception):
    def __init__(self, nome_cidade, message="Non rexistrada en días_chuvia"):
        self.nome_cidade = nome_cidade
        self.message = message
        super().__init__(self.message)
    def __str__(self):
        return f"Cidade: {self.nome_cidade}. {self.message}"

días_chuvia = {'Copenhague': 251, 'Oslo': 232, 'Londres': 226, 'Bruselas': 226, 'Ámsterdam': 214, 'Helsinki': 203,\
               'Estocolmo': 164, 'Viena': 164, 'Praga': 164, 'Santiago de Compostela': 140, 'Dublín': 139}
días_chuvia_lower = {C.lower(): V for C, V in días_chuvia.items()}

while True:
    try:
        print("** CONSULTA DA CHUVIA ANUAL EN EUROPA **")
        print("1 - Consulta por cidade")
        print("2 - Consulta por nº de días de chuvia")
        print("3 - Finalizar programa")
        print("-------------------------------------")
        op = int(input("Opción ->"))
        
        if (op == 1):
            cidade = input("Introduza a cidade:")
            if cidade.lower() not in días_chuvia_lower.keys():
                raise CidadeNonRexistrada(cidade)
            else:
                print(f"O nº de días de chuvia anuais en {cidade} é de {días_chuvia_lower[cidade.lower()]}")
        elif (op == 2):
            n_días = int(input("Introduza o nº de días:"))
            if  n_días < 0:
                raise ValueError("Non se permite un nº de días negativo")
            elif n_días not in días_chuvia.values():
                raise ValueError("Non se atoparon cidades con ese número de días de chuvia")
            else:
                for C, V in días_chuvia.items():
                    if n_días == V:
                        print(C)          
        elif (op == 3):
            print("Fin de programa")
            break
        else:
            raise ValueError("Opción non válida (teclee 1, 2, ou 3)")
    
    except (ValueError,CidadeNonRexistrada) as error:
        print(error)

```

• Se consultamos por nome de produto, debe de contemplarse o caso de que o mesmo non estea no dicionario. Nese caso lanzaremos unha excepción personalizada de nome ProdutoNonRexistrado na que se construirá unha mexase

```python

inventario = {'Refresco cola': 0.9, 'Pizza': 3.29, 'Helado': 1.52, 'Patatas onduladas': 1.91, 'Salsa barbacoa': 2.09, 'Panecillos': 0.9}

class ProdutoNonRexistrado(Exception):
    def __init__(self, prod, mens="Non rexistrado no inventario."):
        self.mens = mens
        self.prod = prod
        super().__init__(self.mens)
        
    def __str__(self):
        return f"Produto: {self.prod}. {self.mens}"
        
while True:
    try:
        print("**CONSULTA DE PRODUTOS DE SUPERMERCADO**")
        print("1 - Consulta por nome de produto")
        print("2 - Consulta por prezo")
        print("3 - Finalizar programa")
        print("----------------------------------------")
        option = int(input("Introduza a opción: "))
        print(f"Opción -> {option}")
        
        if option == 1:
            try:
                nome = input("Introduza o nome do produto: ")
                print(f"Introduza o nome do produto: {nome}")
                if nome.lower() not in str(inventario.keys()).lower():
                    raise ProdutoNonRexistrado(nome)
                else:
                    for i in inventario:
                        if nome.lower() == i.lower():
                            print(f"O prezo do produto {nome.lower()} é de {inventario[i]} euros.")
                
            except ProdutoNonRexistrado as error:
                print(error)
              
        if option == 2:
            try:
                prezo = input("Introduza o prezo: ")
                print(f"Introduza o prezo: {prezo}\n")
                prezo = float(prezo)
            except ValueError as error:
                print("O prezo debe ser un valor numérico")
            else:
                try:
                    if prezo <= 0:
                        raise ValueError("Non se permite un prezo cero ou negativo")
                    elif prezo not in inventario.values():
                        raise ValueError ("Non se atoparon produtos con ese prezo")
                    else:
                        for i in inventario:
                            if prezo == inventario[i]:
                                print(i)    
                                
                except ValueError as e:
                    print(e)            
        
        if option == 3:
            print("Fin de programa")
            break
        
        else:
            if isinstance(option, int) and option not in (1,2,3):
                print("Opción non válida (teclee 1, 2 ou 3)")
            else:
                raise ValueError()       
        
    except (ValueError, TypeError) as error:
        print(error)
    
    
    
    
    
    
    

```
