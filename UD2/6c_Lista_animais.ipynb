## Exercicio 1

Supón que temos unha tupla cos nomes de varios animais:

- **animales** = ('perro','gato','león','perro','tigre','delfín','gato','león','perro')

Recorre a tupla mediante un "for" e contabiliza o número de animais que temos de cada especie nas variables **conta_perro**, **conta_gato**, **conta_león**, **conta_tigre** e **conta_delfín**. Ao final mostra por pantalla os valores obtidos en ditas variables.

```python
animales = ('perro','gato','león','perro','tigre','delfín','gato','león','perro')
conta_perro = 0
conta_gato = 0
conta_leon = 0
conta_tigre = 0
conta_delfin = 0

for animal in animales:
    if 'perro' in animal:
        conta_perro += 1
    elif 'gato' in animal:
        conta_gato +=1
    elif 'león' in animal:
        conta_leon +=1
    elif 'tigre' in animal:
        conta_tigre +=1
    elif 'delfín' in animal:
        conta_delfin +=1
        
print(f"Perros: {conta_perro}, gatos: {conta_gato}, leones: {conta_leon}, tigres: {conta_tigre}, delfines: {conta_delfin}.")
```

## Exercicio 2

Supón que temos unha tupla cos nomes de varios animais:

- **animales** = ('perro','gato','león','perro','tigre','delfín','gato','león','perro')

Usa o método *count()* para contabilizar o número de animais que temos de cada especie e garda dito dato nas variables **conta_perro**, **conta_gato**, **conta_león**, **conta_tigre** e **conta_delfín**. Ao final mostra por pantalla os valores obtidos en ditas variables.

```python
animales = ('perro','gato','león','perro','tigre','delfín','gato','león','perro')
conta_perro = animales.count('perro')
conta_gato = animales.count('gato')
conta_leon = animales.count('león')
conta_tigre = animales.count('tigre')
conta_delfin = animales.count('delfín')

print(f"Perros: {conta_perro}, gatos: {conta_gato}, leones: {conta_leon}, tigres: {conta_tigre}, delfines: {conta_delfin}.")
```

## Exercicio 3

Supón que temos unha tupla cos nomes de varios animais:

- **animales** = ('perro','gato','león','perro','tigre','delfín','gato','león','perro')

Crea un diccionario chamado **ejemplares** no que as claves sexan os nomes dos animais e os valores estean inicialmente postos a cero.

Despois recorre a tupla mediante un "for" para contabilizar o número de animais que temos de cada especie e vai actualizando os valores do diccionario. Ao final mostra por pantalla o diccionario.

```python
animales = ('perro','gato','león','perro','tigre','delfín','gato','león','perro')

ejemplares = {'perro': 0, 'gato': 0, 'león': 0, 'tigre': 0, 'delfín': 0}

for animal in animales:
    if 'perro' in animal:
        ejemplares['perro'] += 1
    elif 'gato' in animal:
        ejemplares['gato'] += 1
    elif 'león' in animal:
        ejemplares['león'] += 1
    elif 'tigre' in animal:
        ejemplares['tigre'] += 1
    elif 'delfín' in animal:
        ejemplares['delfín'] += 1

print(f"Perros: {ejemplares['perro']}, gatos: {ejemplares['gato']}, leones: {ejemplares['león']}, tigres: {ejemplares['tigre']}, delfines: {ejemplares['delfín']}.")
print(f"Diccionario: {ejemplares}")

```
