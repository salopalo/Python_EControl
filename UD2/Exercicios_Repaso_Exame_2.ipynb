Escribe por pantalla os nomes dos alumnos xunto coa súa nota media. Percorre o diccionario de forma que os nomes se amosen por orden alfabético.

```python
dic_alumnos = {"Mateo":(5.7,8.0,9.2), "Noelia":(7.5,6.3), "Manuel Jesús":(2.3,6.3,1.6,4.5)}

for clave, valor in sorted(dic_alumnos.items()):
    print(f"Alumno: {clave} Notas: {round(sum(valor)/len(valor),2)}")
    
    
    
```

```python
# solucion

dic_alumnos = {"Mateo":(5.7,8.0,9.2), "Noelia":(7.5,6.3), "Manuel Jesús":(2.3,6.3,1.6,4.5)}

for nome in sorted(dic_alumnos.keys()):
    total = 0
    conta = 0
    for nota in dic_alumnos[nome]:
        total += nota
        conta += 1
    print(f"{nome}: {(total/conta):.1f}")

#Esta sería unha alternativa máis compacta e sen necesidade de usar as variables total e conta
#for nome, lista_notas in sorted(dic_alumnos.items()):
#    print(f"{nome}: {(sum(lista_notas)/len(lista_notas)):.1f}")
```

## Exercicio 5

Dadas as seguintes listas de países:

 **América_Norte = ["Canada", "Estados Unidos", "México"]**

Convirte ambas as dúas listas a tuplas e despois crea unha nova tupla chamada **América** que sexa a concatenación das novas tuplas e imprímea por pantalla.

Volve a convertir a listas as variables de **América_Norte** e **América_Sur**, e engade os países de "Guatemala" e "Honduras". Percorre a continuación as dúas listas cun iterable ZIP e mostra os nomes dos países que comecen pola letra "C" e indicando si son do norte ou do sur.


```python
América_Norte = ["Canada", "Estados Unidos", "México"]
América_Sur = ["Brasil", "Argentina", "Colombia", "Uruguay", "Chile"]

Am_N = tuple(América_Norte)
Am_S = tuple(América_Sur)

América = Am_N + Am_S
#América = (*Am_N, *Am_S) #utilizar desempaquetado, con Python 3.5+
print(América)

Am_N = list(Am_N)
print(Am_N)
Am_S = list(Am_S)
print(Am_S)
Am_N.append("Guatemala")
Am_N.append("Honduras")

for norte, sur in zip(Am_N, Am_S):
    if norte[0]=="C":
        print(f"Norte: {norte}")
    if sur[0]=="C":
        print(f"Sur: {sur}")
        



```
