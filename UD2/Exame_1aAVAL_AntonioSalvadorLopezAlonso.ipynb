## Exercicio 1

Dada a seguinte lista na que se indica a facturación en euros dunha empresa ao longo da semana (de luns a domingo):

 **facturación = [530.20, 770.95, 600.24, 870.50, 685.30, 320.82, 422.90]** 

Calcula mediante ***"slicing"*** e máis as funcións intrínsecas ***sum()*** e ***len()*** a media de facturación de luns a venres e a media de facturación da fin de semana.

```python
facturación = [530.20, 770.95, 600.24, 870.50, 685.30, 320.82, 422.90]

media_l_v = sum(facturación[:5])/len(facturación[:5])
media_finde = sum(facturación[5:])/len(facturación[5:])
print(f"Media facturación Luns a Venres: {media_l_v:.2f}€.")
print(f"Media facturación fin de semana: {media_finde:.2f}€.")
```

## Exercicio 3

Dada a seguinte lista de palabras:

 **palabras = ["mesa", "encerado", "teclado", "pantalla", "televisor"]** 

Crea unha nova lista por comprensión chamada **palabras_varias_e** que conteña as palabras da lista orixinal que teñan máis dun carácter 'e'.

Percorre a lista **palabras** mostrando por pantalla o seu contido de tal maneira que substitúas en cada palabra o carácter 'e' polo carácter 'E'. Utiliza bucles **for** aniñados e o condicional **if-else**.


```python
palabras = ["mesa", "encerado", "teclado", "pantalla", "televisor"]

palabras_varias_e = [pal for pal in palabras if pal.count("e")>=2]
print(palabras_varias_e)

for p in palabras:
    if "e" in p:
        p.replace("e","E")
        print (p)
    else:
        print(p)
        
```

## Exercicio 4

Dada a seguinte lista na que se indican os resultados obtidos por un determinado cabalo de carreiras nas últimas competicións disputadas, e que está ordenada empezando polos seus resultados máis antigos:

 **resultados = [1, 5, 2, 1, 2, 3, 8, 1, 1, 4, 1]** 

Engade á lista o último resultado acadado (un 3º posto). Calcula mediante o método ***count()*** o número de carreiras nas que resultou vencedor. Calcula tamén mediante un ***for*** o número de carreiras nas que rematou peor do 3º posto.



```python
resultados = [1, 5, 2, 1, 2, 3, 8, 1, 1, 4, 1]

resultados.append(3)
print(resultados)

print(f"O cabalo venceu {resultados.count(1)} veces.")

podium = 0
for r in resultados:
    if r > 3:
        podium += 1
    
print(f"O cabalo rematou peor do 3º posto {podium} veces.")

```

 **hipódromos = ["La Zarzuela", "La Zarzuela", "Andalucía", "Lasarte", "La Zarzuela", "Lasarte"]** 

E a lista dos últimos resultados acadados:

 **resultados = [7, 2, 3, 1, 1, 2]**

Crea unha nova lista por comprensión usando ZIP cuxos elementos sexan tuplas que conteñan pares hipódromo-resultado, de tal maneira que o resultado final sexa o seguinte:

 **nova_lista = [('La Zarzuela', 7), ('La Zarzuela', 2), ('Andalucía', 3), ('Lasarte', 1), ('La Zarzuela', 1), ('Lasarte', 2)]**

## Exercicio 6

A partir das seguintes listas que conteñen o nome de varios tenistas e o nº de títulos de Grand Slam de cada un deles:

 **tenistas = ["Pete Sampras", "Rafael Nadal", "Roger Federer", "Novak Djokovic", "Roy Emerson"]**

 **títulos = [14, 22, 20, 24, 12]**

Crea un dicionario por comprensión chamado **ranking_GS** no que as claves sexan os nomes dos tenistas e os valores o nº de títulos de GS correspondente a cada un deles. Utilizando a función **max()** mostra por pantalla o tenista con máis nº de GS e a cantidade deles que gañou.



```python
tenistas = ["Pete Sampras", "Rafael Nadal", "Roger Federer", "Novak Djokovic", "Roy Emerson"]
títulos = [14, 22, 20, 24, 12]

ranking_GS = {nom:tit for nom,tit in zip(tenistas,títulos)}
print(ranking_GS)

print(f"O tenista con maior cantidade de Grand Slams é {max(ranking_GS.items(), key=lambda x:x[1])[0]}, cun total de {max(ranking_GS.items(), key=lambda x:x[1])[1]} victorias.")
```

## Exercicio 9

Utilizando o dicionario do exercicio anterior:

 **ranking_GS = {'Pete Sampras': 14, 'Rafael Nadal': 22, 'Roger Federer': 20, 'Novak Djokovic': 24, 'Roy Emerson': 12, 'Carlos Alcaraz': 7}**
 
Mostra por pantalla o tenista que teña un número de GS igual a 20 obtendo a lista de nomes e a de títulos e usando o método ***index()***. Crea unha lista por comprensión chamada **listilla** que conteña os nomes dos tenistas con máis de 15 Grand Slams gañados e cuxo nome comece por 'R'.



```python
ranking_GS = {'Pete Sampras': 14, 'Rafael Nadal': 22, 'Roger Federer': 20, 'Novak Djokovic': 24, 'Roy Emerson': 12, 'Carlos Alcaraz': 7}

nomes = list(ranking_GS.keys())
titulos = list(ranking_GS.values())

print(f"O tenista con 20 títulos de Grand Slam é {nomes[titulos.index(20)]}.")

listilla = [nom for nom,tit in zip(nomes,titulos) if tit > 15 and nom[0] == "R"]
print(listilla)
```

## Exercicio 10

Dado o seguinte dicionario no que se indican as notas obtidas por distintos alumnos:

 **grupo_clase = {'Pepe': (1,2,3.7,5), 'Juan': (9,3,4,5), 'Marcos': (2,8,7,8,5)}**
 
Escribe por pantalla o nome dos alumnos xunto coa súa nota mínima. Percorre o dicionario de forma que os nomes se amosen por orden alfabético.

Escribe por pantalla o nome dos alumnos xunto coa súa nota máxima. Percorre o dicionario de forma que os nomes se amosen por orden de nota máxima (de maior a menor).



```python
grupo_clase = {'Pepe': (1,2,3.7,5), 'Juan': (9,3,4,5), 'Marcos': (2,8,7,8,5)}

for nom,nota in sorted(grupo_clase.items()):
    print(f"Nome: {nom}. Nota media: {round(sum(nota)/len(nota),2)}")
    
print("---------------------------------------------------------------------")

for nom,nota in sorted(grupo_clase.items(), key=lambda x:x[1], reverse=True):
    print(f"Nome: {nom}. Nota máxima: {max(nota)}")
    
print("---------------------------------------------------------------------")

    

```
