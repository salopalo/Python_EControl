## Exercicio 1

Dadas as seguintes listas de frutas e de lugares de orixe:

 **frutas = ['manzana', 'plátano', 'kiwi', 'pera', 'fresa']**
 
 **orixe = ['Asturias', 'Canarias', 'Galicia', 'La Rioja', 'Huelva']**

Percorre as dúas listas mediante un iterable ZIP para escribir por pantalla o nome da fruta e o seu orixe.


```python
frutas = ['manzana', 'plátano', 'kiwi', 'pera', 'fresa']
orixe = ['Asturias', 'Canarias', 'Galicia', 'La Rioja', 'Huelva']


for fruta, lugar in zip(frutas, orixe):
    print(f"La {fruta} es de {lugar}")
```

 **habitantes = [ 98179, 292374, 244700, 103756, 97211, 82828]**

Crea un diccionario por comprensión chamado **poboación** no que as claves sexan os nomes das cidades e os valores sexan o dato correspondente á poboación de cada cidade. Para crear o diccionario por comprensión utiliza as dúas listas: **cidades** e **habitantes** e un **"for"** que itere desde 0 ata 5

Fai o mesmo de antes pero usando ZIP.

```python
cidades = ["Santiago de Compostela", "Vigo", "A Coruña", "Ourense", "Lugo", "Pontevedra"]
habitantes = [ 98179, 292374, 244700, 103756, 97211, 82828]

poboacion = {cidades[i]:habitantes[i] for i in range(len(cidades))}
print(poboacion)

poboacion = {cidade: hab for cidade, hab in zip(cidades,habitantes)}
print(poboacion)
```

Mostra por pantalla o nome da cidade que teña 97211 habitantes. Para atopala recorre o diccionario cun **"for"** ata dar co primeiro valor que coincida co dato. Atopa unha solución alternativa obtendo as listas de claves e de valores e usando o método index() da lista de valores.

Utilizando a función ***min()*** mostra por pantalla a cidade co menor nº de habitantes e o seu valor.

```python
poboacion = {'Santiago de Compostela': 98179, 'Vigo': 292374, 'A Coruña': 244700, 'Ourense': 103756, 'Lugo': 97211, 'Pontevedra': 82828}

# Recorrer el diccionario con for
for cidade, hab in poboacion.items():
    if hab == 97211:
        print(f"La ciudad con 97211 habitantes es: {cidade}")
        break
    
    
# Obtener listas de claves y valores
claves = list(poboacion.keys())
valores = list(poboacion.values())

# Buscar el índice del valor 97211
indice = valores.index(97211)

# Obtener la ciudad correspondiente
print(f"La ciudad con 97211 habitantes es: {claves[indice]}")



# Encontrar la ciudad con menos habitantes
cidade_min = min(poboacion, key=poboacion.get)
habitantes_min = poboacion[cidade_min]

print(f"La ciudad con menor población es: {cidade_min} con {habitantes_min} habitantes")
```

## Exercicio 6

A partir da listas seguintes, que conteñen o nome de varios produtos e os seus prezos:  

 **produtos = ["Refresco cola", "Pizza", "Xeado", "Patatas Onduladas", "Salsa Barbacoa"]**
 
 **prezos = [ 0.90, 3.29, 1.52, 1.91, 2.09]**

Crea un diccionario por comprensión chamado **inventario** no que as claves sexan os nomes dos produtos e os valores sexan os prezos correspondentes ao produto. Para crear o diccionario por comprensión utiliza as dúas listas: **produtos** e **prezos** e un ZIP.

```python
produtos = ["Refresco cola", "Pizza", "Xeado", "Patacas Onduladas", "Salsa Barbacoa"]
prezos = [ 0.90, 3.29, 1.52, 1.91, 2.09]

inventario = {prod:prez for prod, prez in zip(produtos, prezos)}
print(inventario)


# ordenar por precio descendente
print(sorted(inventario.items(), key=lambda item: item[1], reverse=True))
# formato diccionario
print(dict(sorted(inventario.items(), key=lambda item: item[1], reverse=True)))
# imprimir productos con precios mayor de 1.50€ en orden ascendente
print(sorted({prod:prez for prod, prez in zip(produtos,prezos) if prez > 1.5}.items(), key=lambda item: item[1], reverse=False))
```
