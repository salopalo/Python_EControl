## Exercicio 1

Dada a seguinte lista na que se indica a cotización media mensual do Euríbor ao longo do ano 2025:

 **euríbor = [2.525, 2.407, 2.398, 2.143, 2.081, 2.081, 2.079, 2.114, 2.172, 2.187, 2.217]** 

Calcula mediante un **"for"** e unha variable chamada **"total"** (que faga de acumulador) a media do segundo trimestre (abril-xuño) e mostra o resultado por pantalla.

```python


euríbor = [2.525, 2.407, 2.398, 2.143, 2.081, 2.081, 2.079, 2.114, 2.172, 2.187, 2.217]
total = 0

for e in range(len(euríbor)):
    if 2 < e < 6:
        total += euríbor[e]
print (f"Media segundo trimestre: {total/3:.3f}.")

```

Calcula mediante ***"slicing"*** e máis as funcións intrínsecas **sum** e **len** a media do terceiro trimestre (xullo-setembro) e mostra o resultado por pantalla.

Mostra por pantalla o nome do mes (abreviado en inglés) que teña por euríbor o dato 2.114. Para atopalo recorre o diccionario cun **"for"** ata dar co primeiro valor que coincida co dato.

Utilizando a función ***max()*** mostra por pantalla o mes no que o euríbor estivo máis alto e o seu valor.

```python
for e in euribor_2025:
    if euribor_2025[e] == 2.114:
        print (f"O primeiro mes co euríbor a 2.114 foi: {e}.")
        break
    
mes_max = max(euribor_2025, key=euribor_2025.get)
print(f"O mes co valor de euríbor máis alto foi: {mes_max} -> {euribor_2025[mes_max]}")
```
